```python
def cleanup():
    import gc
    gc.collect()
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

def nrmse(a,b,eps=1e-12):
    return float((a.float()-b.float()).pow(2).mean().sqrt()/
                 (a.float().pow(2).mean().sqrt()+eps))

def cosine_flat(a,b,eps=1e-12):
    a=a.detach().float().reshape(-1)
    b=b.detach().float().reshape(-1)
    return float(torch.dot(a,b)/(a.norm()*b.norm()+eps))

@torch.no_grad()
def qint8_rows(x):
    xf=x.float()
    s=(xf.abs().amax(dim=1,keepdim=True)/127.0).clamp_min(1e-8)
    q=torch.round(xf/s).clamp(-127,127).to(torch.int8)
    return q,s.to(torch.float16)

@torch.no_grad()
def dqint8_rows(q,s,dtype=torch.float16):
    return q.to(dtype)*s.to(dtype)

def row_int8_bytes(rows,cols):
    return rows*cols + rows*2
```

```python
SUPPORT_CPU = {"language": [], "code": []}
with torch.no_grad():
    for domain in ["language", "code"]:
        for sample in range(len(PANELS["support"][domain])):
            ids = PANELS["support"][domain][sample].unsqueeze(0).to(DEVICE)
            pos = torch.arange(SEQ_LEN, device=DEVICE, dtype=torch.long).unsqueeze(0)
            H = model.transformer.wte(ids) + model.transformer.wpe(pos)
            H = H.to(model.transformer.h[0].ln_1.weight.dtype)
            states = [H.squeeze(0).cpu()]
            for l in range(NLAYERS):
                block = model.transformer.h[l]
                H = H + block.attn(block.ln_1(H), use_cache=False)[0]
                H = H + block.mlp(block.ln_2(H))
                states.append(H.squeeze(0).cpu())
            SUPPORT_CPU[domain].append(states)


def parse_x_mode(mode):
    m=re.fullmatch(r"x8g(\d+)",mode)
    if m is None:
        raise ValueError(mode)
    return int(m.group(1))


@torch.no_grad()
def native_ln1_xhat(H,layer):
    block=model.transformer.h[layer]
    N=block.ln_1(
        H.to(block.ln_1.weight.dtype)
    ).float()

    gamma=block.ln_1.weight.detach().float()
    beta=block.ln_1.bias.detach().float()

    if float(gamma.abs().min())<=1e-6:
        raise RuntimeError({
            "ln1_gamma_too_small":{
                "layer":layer,
                "min_abs_gamma":float(gamma.abs().min()),
            }
        })

    return N,(N-beta)/gamma


@torch.no_grad()
def pack_saved_xhat_base(xhat,mode):
    group=parse_x_mode(mode)

    shape=tuple(xhat.shape)
    rows=int(np.prod(shape[:-1]))
    cols=shape[-1]

    assert cols==HIDDEN
    assert cols%group==0

    x=xhat.float().reshape(
        rows,
        cols//group,
        group,
    )

    scale=(
        x.abs().amax(dim=-1,keepdim=True)/
        127.0
    ).clamp_min(1e-8)

    q=torch.round(
        x/scale
    ).clamp(
        -127,
        127,
    ).to(torch.int8).contiguous()

    s=scale.squeeze(-1).to(torch.float16).contiguous()

    return q,s


@torch.no_grad()
def unpack_saved_xhat_base(q,s,shape,mode):
    return (
        q.float()*
        s.float().unsqueeze(-1)
    ).reshape(shape)


@torch.no_grad()
def pack_two_bit_stream(codes):
    flat=codes.reshape(-1).to(torch.uint8)
    assert flat.numel()%4==0
    assert bool(((flat<=3).all()).item())

    g=flat.reshape(-1,4)

    packed=(
        g[:,0]
        |
        (g[:,1] << 2)
        |
        (g[:,2] << 4)
        |
        (g[:,3] << 6)
    ).to(torch.uint8).contiguous()

    return packed


@torch.no_grad()
def unpack_two_bit_stream(packed,shape):
    numel=int(np.prod(shape))
    assert numel%4==0
    assert packed.numel()==numel//4

    p=packed.reshape(-1)

    out=torch.empty(
        numel,
        dtype=torch.uint8,
        device=packed.device,
    )

    out[0::4]=p & 3
    out[1::4]=(p >> 2) & 3
    out[2::4]=(p >> 4) & 3
    out[3::4]=(p >> 6) & 3

    return out.reshape(shape)


@torch.no_grad()
def residual_to_two_bit_codes(
    residual,
    scale,
    mode,
):
    group=parse_x_mode(mode)

    shape=tuple(residual.shape)
    rows=int(np.prod(shape[:-1]))
    cols=shape[-1]
    ng=cols//group

    rv=residual.float().reshape(
        rows,
        ng,
        group,
    )

    sv=scale.float().reshape(
        rows,
        ng,
        1,
    )

    normalized=(rv/sv).clamp(
        -0.5,
        0.5,
    )

    codes=torch.floor(
        (normalized+0.5)*4.0
    ).clamp(
        0,
        3,
    ).to(torch.uint8)

    return codes.reshape(shape)


@torch.no_grad()
def pack_saved_xhat_subbin2(xhat,mode):
    q,s=pack_saved_xhat_base(
        xhat,
        mode,
    )

    base=unpack_saved_xhat_base(
        q,
        s,
        xhat.shape,
        mode,
    )

    residual=xhat.float()-base

    codes=residual_to_two_bit_codes(
        residual,
        s,
        mode,
    )

    packed2=pack_two_bit_stream(
        codes
    )

    assert packed2.numel()==xhat.numel()//4

    del base,residual,codes

    return q,s,packed2


@torch.no_grad()
def apply_xhat_residual_view(
    base,
    scale,
    packed2,
    mode,
    fine,
):
    group=parse_x_mode(mode)

    shape=tuple(base.shape)
    rows=int(np.prod(shape[:-1]))
    cols=shape[-1]
    ng=cols//group

    out=base
    codes=unpack_two_bit_stream(
        packed2,
        shape,
    )

    ov=out.reshape(rows,ng,group)
    cv=codes.reshape(rows,ng,group)
    sv=scale.float().reshape(rows,ng)

    chunk_groups=8

    for start in range(0,ng,chunk_groups):
        end=min(ng,start+chunk_groups)

        if fine:
            # k=0,1,2,3 -> -3/8,-1/8,+1/8,+3/8
            corr=cv[:,start:end,:].to(torch.float32)
            corr.mul_(2.0).sub_(3.0)
            corr.mul_(1.0/8.0)
        else:
            # Coarse parent one-bit view, derived from the same two-bit code.
            corr=(cv[:,start:end,:]>=2).to(torch.float32)
            corr.mul_(2.0).sub_(1.0)
            corr.mul_(0.25)

        corr.mul_(
            sv[:,start:end].unsqueeze(-1)
        )

        ov[:,start:end,:].add_(corr)

        del corr

    del codes

    return out


@torch.no_grad()
def unpack_saved_xhat_coarse(
    q,s,packed2,shape,mode,
):
    base=unpack_saved_xhat_base(
        q,s,shape,mode
    )

    return apply_xhat_residual_view(
        base,
        s,
        packed2,
        mode,
        fine=False,
    )


@torch.no_grad()
def unpack_saved_xhat_fine(
    q,s,packed2,shape,mode,
):
    base=unpack_saved_xhat_base(
        q,s,shape,mode
    )

    return apply_xhat_residual_view(
        base,
        s,
        packed2,
        mode,
        fine=True,
    )


def x_mode_bytes_per_sequence(mode):
    group=parse_x_mode(mode)

    return int(
        SEQ_LEN*HIDDEN+
        SEQ_LEN*(HIDDEN//group)*2+
        SEQ_LEN*2
    )


def subbin2_bytes_per_sequence():
    return int(
        SEQ_LEN*HIDDEN*2//8
    )


@torch.no_grad()
def two_bit_stream_exactness_audit():
    codes=torch.arange(
        0,
        4,
        dtype=torch.uint8,
        device=DEVICE,
    ).repeat(
        4096
    )

    packed=pack_two_bit_stream(
        codes
    )

    decoded=unpack_two_bit_stream(
        packed,
        codes.shape,
    )

    return {
        "num_codes":int(codes.numel()),
        "packed_bytes":int(packed.numel()),
        "exact":bool(torch.equal(codes,decoded)),
    }


@torch.no_grad()
def attention_multires_codec_audit():
    layer=10
    mode=FROZEN_ATTN_X_MODES[layer]

    H=SUPPORT_CPU["language"][0][layer].to(DEVICE).unsqueeze(0)

    _,xhat=native_ln1_xhat(
        H,
        layer,
    )

    q,s,p2=pack_saved_xhat_subbin2(
        xhat,
        mode,
    )

    base=unpack_saved_xhat_base(
        q,s,xhat.shape,mode
    )

    residual=xhat-base

    # Direct parent one-bit logical view.
    group=parse_x_mode(mode)
    rows=int(np.prod(xhat.shape[:-1]))
    ng=HIDDEN//group

    parent=base.clone().reshape(rows,ng,group)
    rv=residual.reshape(rows,ng,group)
    sv=s.float().reshape(rows,ng)

    parent.add_(
        torch.where(
            rv>=0,
            torch.ones_like(rv),
            -torch.ones_like(rv),
        )*
        sv.unsqueeze(-1)*
        0.25
    )

    parent=parent.reshape(xhat.shape)

    coarse=unpack_saved_xhat_coarse(
        q,s,p2,xhat.shape,mode
    )

    fine=unpack_saved_xhat_fine(
        q,s,p2,xhat.shape,mode
    )

    result={
        "layer":layer,
        "mode":mode,
        "two_bit_stream_audit":two_bit_stream_exactness_audit(),
        "coarse_parent_identity_maxabs":float(
            (coarse-parent).abs().max()
        ),
        "base_xhat_nrmse":nrmse(
            xhat,base
        ),
        "coarse_xhat_nrmse":nrmse(
            xhat,coarse
        ),
        "fine_xhat_nrmse":nrmse(
            xhat,fine
        ),
        "fine_over_coarse_nrmse_ratio":float(
            (xhat-fine).norm()/
            ((xhat-coarse).norm()+1e-12)
        ),
        "two_bit_sidecar_bytes_per_sequence":subbin2_bytes_per_sequence(),
        "expected_sidecar_bytes":int(xhat.numel()*2//8),
        "actual_sidecar_bytes":int(p2.numel()),
    }

    result["passed"]=bool(
        result["two_bit_stream_audit"]["exact"]
        and
        result["coarse_parent_identity_maxabs"]==0.0
        and
        result["actual_sidecar_bytes"]==result["expected_sidecar_bytes"]
        and
        result["fine_xhat_nrmse"]<
        result["coarse_xhat_nrmse"]
    )

    return result


ATTN_MULTIRES_CODEC_AUDIT=attention_multires_codec_audit()

(OUT/"attention_two_bit_multires_codec_audit.json").write_text(
    json.dumps(
        ATTN_MULTIRES_CODEC_AUDIT,
        indent=2,
    ),
    encoding="utf-8",
)

print(
    json.dumps(
        ATTN_MULTIRES_CODEC_AUDIT,
        indent=2,
    )
)

if not ATTN_MULTIRES_CODEC_AUDIT["passed"]:
    raise RuntimeError({
        "attention_multires_codec_audit_failed":ATTN_MULTIRES_CODEC_AUDIT
    })
```

```python
def attention_params(layer):
    block=model.transformer.h[layer]

    return [
        block.ln_1.weight,
        block.ln_1.bias,
        block.attn.c_attn.weight,
        block.attn.c_attn.bias,
        block.attn.c_proj.weight,
        block.attn.c_proj.bias,
    ]


def set_param_flags(params,enabled=True):
    old=[
        bool(
            p.requires_grad
        )
        for p in params
    ]

    for p in params:
        p.requires_grad_(
            enabled
        )

    return old


def restore_param_flags(params,flags):
    for p,f in zip(
        params,
        flags,
    ):
        p.requires_grad_(
            f
        )


def deterministic_attention_upstream(
    layer,
    domain,
    phase,
):
    gen=torch.Generator(
        device=DEVICE
    )

    gen.manual_seed(
        SEED+
        31013*layer+
        191*sum(
            ord(c)
            for c in domain
        )+
        227*sum(
            ord(c)
            for c in phase
        )
    )

    g=torch.randn(
        (
            SEQ_LEN,
            HIDDEN,
        ),
        generator=gen,
        device=DEVICE,
        dtype=torch.float32,
    )

    return g/(
        g.norm()+
        1e-12
    )


def native_attention_case(
    H0,
    layer,
    upstream,
):
    block=model.transformer.h[
        layer
    ]

    params=attention_params(
        layer
    )

    old=set_param_flags(
        params,
        True
    )

    try:
        with torch.enable_grad():
            H=(
                H0.detach()
                .clone()
                .to(
                    block.ln_1.weight.dtype
                )
                .requires_grad_(
                    True
                )
            )

            N=block.ln_1(
                H
            )

            A=block.attn(
                N,
                use_cache=False,
            )[0]

            out=H+A

            loss=(
                out.float()*
                upstream
            ).sum()

            grads=torch.autograd.grad(
                loss,
                [
                    H
                ]+
                params,
                retain_graph=False,
                create_graph=False,
                allow_unused=False,
            )

        return (
            out.detach().float(),
            N.detach().float(),
            [
                g.detach().float()
                for g in grads
            ],
        )

    finally:
        restore_param_flags(
            params,
            old
        )


def recompute_attention_from_n(
    N0,
    layer,
    upstream,
):
    block=model.transformer.h[
        layer
    ]

    cattn_w=block.attn.c_attn.weight
    cattn_b=block.attn.c_attn.bias
    cproj_w=block.attn.c_proj.weight
    cproj_b=block.attn.c_proj.bias

    params=[
        cattn_w,
        cattn_b,
        cproj_w,
        cproj_b,
    ]

    old=set_param_flags(
        params,
        True
    )

    try:
        with torch.enable_grad():
            N=(
                N0.detach()
                .clone()
                .to(
                    block.ln_1.weight.dtype
                )
                .requires_grad_(
                    True
                )
            )

            A=block.attn(
                N,
                use_cache=False,
            )[0]

            surrogate=(
                A.float()*
                upstream
            ).sum()

            grads=torch.autograd.grad(
                surrogate,
                [
                    N,
                    cattn_w,
                    cattn_b,
                    cproj_w,
                    cproj_b,
                ],
                retain_graph=False,
                create_graph=False,
                allow_unused=False,
            )

        return (
            A.detach().float(),
            [
                g.detach().float()
                for g in grads
            ],
        )

    finally:
        restore_param_flags(
            params,
            old
        )


@torch.no_grad()
def explicit_ln1_backward_from_xhat(
    dN,
    xhat_saved,
    rstd_saved,
    layer,
    upstream_direct,
):
    block=model.transformer.h[layer]

    gamma=block.ln_1.weight.detach().float()

    dy=dN.float()
    xhat=xhat_saved.float()

    reduce_dims=tuple(
        range(
            dy.ndim-1
        )
    )

    dgamma=(
        dy*
        xhat
    ).sum(
        dim=reduce_dims
    )

    dbeta=dy.sum(
        dim=reduce_dims
    )

    dxhat=dy*gamma

    mean_dxhat=dxhat.mean(
        dim=-1,
        keepdim=True,
    )

    mean_dxhat_xhat=(
        dxhat*
        xhat
    ).mean(
        dim=-1,
        keepdim=True,
    )

    dH_ln=(
        rstd_saved.float()*
        (
            dxhat-
            mean_dxhat-
            xhat*
            mean_dxhat_xhat
        )
    )

    dH=(
        upstream_direct.float()+
        dH_ln
    )

    return (
        dH,
        dgamma,
        dbeta,
    )



@torch.no_grad()
def exact_attention_formula_audit():
    layer=4

    H0=SUPPORT_CPU[
        "language"
    ][0][layer].to(
        DEVICE
    ).unsqueeze(0)

    block=model.transformer.h[layer]

    upstream=deterministic_attention_upstream(
        layer,
        "language",
        "exact_xhat_formula_audit",
    ).unsqueeze(0)

    native_out,N,native_grads=native_attention_case(
        H0,
        layer,
        upstream,
    )

    gamma=block.ln_1.weight.detach().float()
    beta=block.ln_1.bias.detach().float()

    xhat=(
        N-
        beta
    )/gamma

    var=H0.float().var(
        dim=-1,
        unbiased=False,
        keepdim=True,
    )

    rstd=torch.rsqrt(
        var+
        float(
            block.ln_1.eps
        )
    )

    N_reconstructed=(
        xhat*
        gamma+
        beta
    )

    A,recomputed=recompute_attention_from_n(
        N_reconstructed,
        layer,
        upstream,
    )

    dN,cattn_w,cattn_b,cproj_w,cproj_b=recomputed

    dH,lnw,lnb=explicit_ln1_backward_from_xhat(
        dN,
        xhat,
        rstd,
        layer,
        upstream,
    )

    Hnative=H0.to(
        block.ln_1.weight.dtype
    )

    recomputed_out=(
        Hnative+
        A.to(
            Hnative.dtype
        )
    ).float()

    approx=[
        dH,
        lnw,
        lnb,
        cattn_w,
        cattn_b,
        cproj_w,
        cproj_b,
    ]

    cosines=[
        cosine_flat(
            g,
            h
        )
        for g,h in zip(
            native_grads,
            approx,
        )
    ]

    result={
        "layer":layer,
        "forward_identity_nrmse":nrmse(
            native_out,
            recomputed_out,
        ),
        "gradient_cosines":cosines,
        "minimum_gradient_cosine":float(
            min(
                cosines
            )
        ),
    }

    result["passed"]=bool(
        result[
            "forward_identity_nrmse"
        ]<=1e-5
        and
        result[
            "minimum_gradient_cosine"
        ]>=0.9999
    )

    return result


FORMULA_AUDIT=exact_attention_formula_audit()

(OUT/"exact_attention_xhat_custom_backward_formula_audit.json").write_text(
    json.dumps(
        FORMULA_AUDIT,
        indent=2,
    ),
    encoding="utf-8",
)

print(
    FORMULA_AUDIT
)

if not FORMULA_AUDIT[
    "passed"
]:
    raise RuntimeError({
        "attention_formula_audit_failed":FORMULA_AUDIT
    })
```

```python
class SHVLN1AttentionFineViewFunction(torch.autograd.Function):

    @staticmethod
    def forward(
        ctx,
        H,
        ln_w,
        ln_b,
        cattn_w,
        cattn_b,
        cproj_w,
        cproj_b,
        layer_idx,
        mode,
    ):
        layer_idx=int(layer_idx)
        block=model.transformer.h[layer_idx]

        assert FORMAL_ATTN_CONSUMER_VIEW=="fine"
        assert ln_w.data_ptr()==block.ln_1.weight.data_ptr()
        assert ln_b.data_ptr()==block.ln_1.bias.data_ptr()
        assert cattn_w.data_ptr()==block.attn.c_attn.weight.data_ptr()
        assert cattn_b.data_ptr()==block.attn.c_attn.bias.data_ptr()
        assert cproj_w.data_ptr()==block.attn.c_proj.weight.data_ptr()
        assert cproj_b.data_ptr()==block.attn.c_proj.bias.data_ptr()

        with torch.no_grad():
            N=block.ln_1(H)
            A=block.attn(N,use_cache=False)[0]
            out=H+A

            gamma=ln_w.detach().float()
            beta=ln_b.detach().float()
            xhat=(N.float()-beta)/gamma

            qx,sx,p2=pack_saved_xhat_subbin2(
                xhat,
                mode,
            )

            var=H.float().var(
                dim=-1,
                unbiased=False,
                keepdim=True,
            )
            rstd=torch.rsqrt(
                var+float(block.ln_1.eps)
            ).to(torch.float16)

        ctx.save_for_backward(
            qx,
            sx,
            p2,
            rstd,
        )
        ctx.param_refs=(
            ln_w,
            ln_b,
            cattn_w,
            cattn_b,
            cproj_w,
            cproj_b,
        )
        ctx.layer_idx=layer_idx
        ctx.x_shape=tuple(xhat.shape)
        ctx.mode=mode
        ctx.h_dtype=H.dtype
        return out


    @staticmethod
    def backward(
        ctx,
        grad_output,
    ):
        qx,sx,p2,rstd=ctx.saved_tensors
        (
            ln_w,
            ln_b,
            cattn_w,
            cattn_b,
            cproj_w,
            cproj_b,
        )=ctx.param_refs

        block=model.transformer.h[
            ctx.layer_idx
        ]
        G=grad_output.float()

        # Formal D2 winner: the nonlinear Attention recomputation consumes
        # the full two-bit fine view.
        xhat_attn=unpack_saved_xhat_fine(
            qx,
            sx,
            p2,
            ctx.x_shape,
            ctx.mode,
        )

        Nhat=(
            xhat_attn*
            ln_w.detach().float()+
            ln_b.detach().float()
        ).to(
            block.ln_1.weight.dtype
        )
        del xhat_attn

        attn_params=[
            cattn_w,
            cattn_b,
            cproj_w,
            cproj_b,
        ]
        old=set_param_flags(
            attn_params,
            True,
        )

        try:
            with torch.enable_grad():
                Nreq=(
                    Nhat.detach()
                    .requires_grad_(True)
                )
                Ahat=block.attn(
                    Nreq,
                    use_cache=False,
                )[0]
                surrogate=(
                    Ahat.float()*
                    G
                ).sum()

                (
                    dN,
                    gcattn_w,
                    gcattn_b,
                    gcproj_w,
                    gcproj_b,
                )=torch.autograd.grad(
                    surrogate,
                    [
                        Nreq,
                        cattn_w,
                        cattn_b,
                        cproj_w,
                        cproj_b,
                    ],
                    retain_graph=False,
                    create_graph=False,
                    allow_unused=False,
                )
        finally:
            restore_param_flags(
                attn_params,
                old,
            )

        del Nhat,Nreq,Ahat,surrogate

        # LN1 uses the same fine normalized-direction view.
        xhat_ln=unpack_saved_xhat_fine(
            qx,
            sx,
            p2,
            ctx.x_shape,
            ctx.mode,
        )

        dH,gln_w,gln_b=explicit_ln1_backward_from_xhat(
            dN.detach().float(),
            xhat_ln,
            rstd,
            ctx.layer_idx,
            G,
        )
        del xhat_ln

        return (
            dH.to(ctx.h_dtype),
            gln_w.to(ln_w.dtype),
            gln_b.to(ln_b.dtype),
            gcattn_w.to(cattn_w.dtype),
            gcattn_b.to(cattn_b.dtype),
            gcproj_w.to(cproj_w.dtype),
            gcproj_b.to(cproj_b.dtype),
            None,
            None,
        )


ATTN_GRAD_NAMES=[
    "block_input",
    "ln1_weight",
    "ln1_bias",
    "cattn_weight",
    "cattn_bias",
    "cproj_weight",
    "cproj_bias",
]
```

```python
def parse_z_mode(mode):
    m=re.fullmatch(r"z([67])g(\d+)",mode)
    if m is None:
        raise ValueError(mode)
    return int(m.group(1)),int(m.group(2))


def fixed_pack_layout(bits):
    if bits==6:
        return 4,3
    if bits==7:
        return 8,7
    raise ValueError(bits)


@torch.no_grad()
def pack_fixed_uint(u,bits):
    u=u.reshape(-1)
    vals_per_word,bytes_per_word=fixed_pack_layout(bits)

    assert u.numel()%vals_per_word==0

    groups=u.to(torch.int64).reshape(-1,vals_per_word)

    word=torch.zeros(
        groups.shape[0],
        dtype=torch.int64,
        device=groups.device,
    )

    for i in range(vals_per_word):
        word |= groups[:,i] << (bits*i)

    out=torch.empty(
        (groups.shape[0],bytes_per_word),
        dtype=torch.uint8,
        device=groups.device,
    )

    for j in range(bytes_per_word):
        out[:,j]=((word>>(8*j))&255).to(torch.uint8)

    return out.reshape(-1).contiguous()


@torch.no_grad()
def unpack_fixed_uint(packed,bits,numel):
    vals_per_word,bytes_per_word=fixed_pack_layout(bits)

    assert numel%vals_per_word==0

    p=packed.reshape(-1,bytes_per_word).to(torch.int64)

    word=torch.zeros(
        p.shape[0],
        dtype=torch.int64,
        device=p.device,
    )

    for j in range(bytes_per_word):
        word |= p[:,j] << (8*j)

    mask=(1<<bits)-1

    vals=[
        (word>>(bits*i))&mask
        for i in range(vals_per_word)
    ]

    return torch.stack(vals,dim=1).reshape(-1)[:numel]


@torch.no_grad()
def pack_saved_z(Z,mode):
    bits,group=parse_z_mode(mode)

    shape=tuple(Z.shape)
    rows=int(np.prod(shape[:-1]))
    cols=shape[-1]

    assert cols==HIDDEN
    assert cols%group==0

    qmax=(1<<(bits-1))-1

    xf=Z.float().reshape(
        rows,
        cols//group,
        group,
    )

    scale=(
        xf.abs().amax(dim=-1,keepdim=True)/
        float(qmax)
    ).clamp_min(1e-8)

    q=torch.round(
        xf/scale
    ).clamp(
        -qmax,
        qmax,
    ).to(torch.int16)

    u=(q.to(torch.int64)+qmax).reshape(-1)

    packed=pack_fixed_uint(
        u,
        bits,
    )

    stored_scale=scale.squeeze(-1).to(torch.float16).contiguous()

    expected=rows*cols*bits//8
    assert packed.numel()==expected

    return packed,stored_scale


@torch.no_grad()
def unpack_saved_z(
    packed,
    scale,
    mode,
    shape,
):
    bits,group=parse_z_mode(mode)

    rows=int(np.prod(shape[:-1]))
    cols=shape[-1]
    qmax=(1<<(bits-1))-1
    numel=rows*cols

    u=unpack_fixed_uint(
        packed,
        bits,
        numel,
    )

    q=(
        u.to(torch.int64)-
        qmax
    ).reshape(
        rows,
        cols//group,
        group,
    )

    Zhat=(
        q.float()*
        scale.float().unsqueeze(-1)
    ).reshape(shape)

    del u,q

    return Zhat


@torch.no_grad()
def residual_to_two_bit_codes_z(
    residual,
    scale,
    mode,
):
    _,group=parse_z_mode(mode)

    shape=tuple(residual.shape)
    rows=int(np.prod(shape[:-1]))
    cols=shape[-1]
    ng=cols//group

    rv=residual.float().reshape(
        rows,
        ng,
        group,
    )

    sv=scale.float().reshape(
        rows,
        ng,
        1,
    )

    normalized=(rv/sv).clamp(
        -0.5,
        0.5,
    )

    codes=torch.floor(
        (normalized+0.5)*4.0
    ).clamp(
        0,
        3,
    ).to(torch.uint8)

    return codes.reshape(shape)


@torch.no_grad()
def pack_saved_z_dual2(
    Z,
    mode,
):
    packed,scale=pack_saved_z(
        Z,
        mode,
    )

    Zhat=unpack_saved_z(
        packed,
        scale,
        mode,
        Z.shape,
    )

    residual=Z.float()-Zhat

    codes=residual_to_two_bit_codes_z(
        residual,
        scale,
        mode,
    )

    packed2=pack_two_bit_stream(
        codes
    )

    assert packed2.numel()==Z.numel()//4

    del Zhat,residual,codes

    return packed,scale,packed2


@torch.no_grad()
def ln2_xhat_from_saved_dual2(
    Zhat,
    scale,
    packed2,
    mode,
    layer,
):
    _,group=parse_z_mode(mode)

    shape=tuple(Zhat.shape)
    rows=int(np.prod(shape[:-1]))
    cols=shape[-1]
    ng=cols//group

    block=model.transformer.h[layer]

    gamma=block.ln_2.weight.detach().float()
    beta=block.ln_2.bias.detach().float()

    if float(gamma.abs().min())<=1e-6:
        raise RuntimeError({
            "ln2_gamma_too_small":{
                "layer":layer,
                "min_abs_gamma":float(gamma.abs().min()),
            }
        })

    xhat=(
        Zhat.float()-
        beta
    )/gamma

    codes=unpack_two_bit_stream(
        packed2,
        shape,
    )

    xv=xhat.reshape(rows,ng,group)
    cv=codes.reshape(rows,ng,group)
    sv=scale.float().reshape(rows,ng)
    gamma_g=gamma.reshape(ng,group)

    chunk_groups=8

    for start in range(0,ng,chunk_groups):
        end=min(ng,start+chunk_groups)

        corr=cv[:,start:end,:].to(torch.float32)
        corr.mul_(2.0).sub_(3.0)
        corr.mul_(1.0/8.0)

        corr.mul_(
            sv[:,start:end].unsqueeze(-1)
        )

        corr.div_(
            gamma_g[start:end,:].unsqueeze(0)
        )

        xv[:,start:end,:].add_(corr)

        del corr

    del codes

    return xhat


def mlp_dual2_bytes_per_sequence(mode):
    bits,group=parse_z_mode(mode)

    base_payload=SEQ_LEN*HIDDEN*bits//8
    scales=SEQ_LEN*(HIDDEN//group)*2
    subbin2=SEQ_LEN*HIDDEN*2//8
    rstd=SEQ_LEN*2

    return int(
        base_payload+
        scales+
        subbin2+
        rstd
    )


@torch.no_grad()
def mlp_dual2_codec_audit():
    layer=2
    mode=FROZEN_MLP_Z_MODES[layer]

    H=SUPPORT_CPU["language"][0][layer].to(DEVICE).unsqueeze(0)
    block=model.transformer.h[layer]

    U=H+block.attn(
        block.ln_1(H),
        use_cache=False,
    )[0]

    Z=block.ln_2(U).float()

    gamma=block.ln_2.weight.detach().float()
    beta=block.ln_2.bias.detach().float()
    xhat_native=(Z-beta)/gamma

    packed,scale,p2=pack_saved_z_dual2(
        Z,
        mode,
    )

    # Parent MLP consumer view.
    Zhat_parent=unpack_saved_z(
        packed,
        scale,
        mode,
        Z.shape,
    )

    Zhat_mlp=unpack_saved_z(
        packed,
        scale,
        mode,
        Z.shape,
    )

    parent_identity=float(
        (Zhat_parent-Zhat_mlp).abs().max()
    )

    # Previous one-bit LN2 logical view reconstructed from two-bit sign.
    _,group=parse_z_mode(mode)
    shape=tuple(Z.shape)
    rows=int(np.prod(shape[:-1]))
    ng=HIDDEN//group

    codes=unpack_two_bit_stream(
        p2,
        shape,
    )

    coarse_z=Zhat_parent.clone().reshape(
        rows,ng,group
    )

    cv=codes.reshape(rows,ng,group)
    sv=scale.float().reshape(rows,ng)

    coarse_corr=(cv>=2).to(torch.float32)
    coarse_corr.mul_(2.0).sub_(1.0)
    coarse_corr.mul_(0.25)
    coarse_corr.mul_(sv.unsqueeze(-1))

    coarse_z.add_(coarse_corr)
    coarse_z=coarse_z.reshape(shape)

    xhat_coarse=(coarse_z-beta)/gamma

    xhat_fine=ln2_xhat_from_saved_dual2(
        Zhat_parent.clone(),
        scale,
        p2,
        mode,
        layer,
    )

    result={
        "layer":layer,
        "mode":mode,
        "mlp_parent_z_identity_maxabs":parent_identity,
        "z_nrmse":nrmse(
            Z,
            Zhat_parent,
        ),
        "coarse_xhat_nrmse":nrmse(
            xhat_native,
            xhat_coarse,
        ),
        "fine_xhat_nrmse":nrmse(
            xhat_native,
            xhat_fine,
        ),
        "fine_over_coarse_xhat_nrmse_ratio":float(
            (xhat_native-xhat_fine).norm()/
            ((xhat_native-xhat_coarse).norm()+1e-12)
        ),
        "two_bit_sidecar_bytes":int(p2.numel()),
        "expected_two_bit_sidecar_bytes":int(Z.numel()*2//8),
        "bytes_per_sequence":mlp_dual2_bytes_per_sequence(mode),
    }

    result["passed"]=bool(
        parent_identity==0.0
        and
        result["two_bit_sidecar_bytes"]==
        result["expected_two_bit_sidecar_bytes"]
        and
        result["fine_xhat_nrmse"]<
        result["coarse_xhat_nrmse"]
    )

    return result


MLP_DUAL2_CODEC_AUDIT=mlp_dual2_codec_audit()

(OUT/"mlp_two_bit_dual_view_codec_audit.json").write_text(
    json.dumps(
        MLP_DUAL2_CODEC_AUDIT,
        indent=2,
    ),
    encoding="utf-8",
)

print(
    json.dumps(
        MLP_DUAL2_CODEC_AUDIT,
        indent=2,
    )
)

if not MLP_DUAL2_CODEC_AUDIT["passed"]:
    raise RuntimeError({
        "mlp_dual2_codec_audit_failed":MLP_DUAL2_CODEC_AUDIT
    })


@torch.no_grad()
def unpack_saved_z_fine(
    packed,
    scale,
    packed2,
    mode,
    shape,
):
    Zhat=unpack_saved_z(packed,scale,mode,shape)
    _,group=parse_z_mode(mode)
    rows=int(np.prod(shape[:-1]))
    cols=shape[-1]
    ng=cols//group

    codes=unpack_two_bit_stream(packed2,shape)
    zv=Zhat.reshape(rows,ng,group)
    cv=codes.reshape(rows,ng,group)
    sv=scale.float().reshape(rows,ng)

    chunk_groups=8
    for start in range(0,ng,chunk_groups):
        end=min(ng,start+chunk_groups)
        corr=cv[:,start:end,:].to(torch.float32)
        corr.mul_(2.0).sub_(3.0)
        corr.mul_(1.0/8.0)
        corr.mul_(sv[:,start:end].unsqueeze(-1))
        zv[:,start:end,:].add_(corr)
        del corr

    del codes
    return Zhat
```

```python
@torch.no_grad()
def explicit_ln2_backward_from_xhat(
    dZ,
    xhat_saved,
    rstd_saved,
    layer,
    upstream_direct,
):
    block=model.transformer.h[layer]

    gamma=block.ln_2.weight.detach().float()

    dy=dZ.float()
    xhat=xhat_saved.float()

    reduce_dims=tuple(
        range(
            dy.ndim-1
        )
    )

    dgamma=(
        dy*
        xhat
    ).sum(
        dim=reduce_dims
    )

    dbeta=dy.sum(
        dim=reduce_dims
    )

    dxhat=dy*gamma

    mean_dxhat=dxhat.mean(
        dim=-1,
        keepdim=True,
    )

    mean_dxhat_xhat=(
        dxhat*
        xhat
    ).mean(
        dim=-1,
        keepdim=True,
    )

    dU_ln=(
        rstd_saved.float()*
        (
            dxhat-
            mean_dxhat-
            xhat*
            mean_dxhat_xhat
        )
    )

    dU=(
        upstream_direct.float()+
        dU_ln
    )

    return dU,dgamma,dbeta





class SHVLN2MLPFineViewFunction(torch.autograd.Function):

    @staticmethod
    def forward(
        ctx,
        U,
        ln_w,
        ln_b,
        cfc_w,
        cfc_b,
        cproj_w,
        cproj_b,
        layer_idx,
        mode,
    ):
        layer_idx=int(layer_idx)
        block=model.transformer.h[layer_idx]

        assert FORMAL_MLP_CONSUMER_VIEW=="fine"
        assert ln_w.data_ptr()==block.ln_2.weight.data_ptr()
        assert ln_b.data_ptr()==block.ln_2.bias.data_ptr()
        assert cfc_w.data_ptr()==block.mlp.c_fc.weight.data_ptr()
        assert cfc_b.data_ptr()==block.mlp.c_fc.bias.data_ptr()
        assert cproj_w.data_ptr()==block.mlp.c_proj.weight.data_ptr()
        assert cproj_b.data_ptr()==block.mlp.c_proj.bias.data_ptr()

        with torch.no_grad():
            Z=block.ln_2(U)
            M=block.mlp(Z)
            out=U+M

            packed,scale,p2=pack_saved_z_dual2(
                Z,
                mode,
            )

            var=U.float().var(
                dim=-1,
                unbiased=False,
                keepdim=True,
            )
            rstd=torch.rsqrt(
                var+float(block.ln_2.eps)
            ).to(torch.float16)

        ctx.save_for_backward(
            packed,
            scale,
            p2,
            rstd,
        )
        ctx.param_refs=(
            ln_w,
            ln_b,
            cfc_w,
            cfc_b,
            cproj_w,
            cproj_b,
        )
        ctx.layer_idx=layer_idx
        ctx.mode=mode
        ctx.z_shape=tuple(Z.shape)
        ctx.u_dtype=U.dtype
        return out


    @staticmethod
    def backward(
        ctx,
        grad_output,
    ):
        packed,scale,p2,rstd=ctx.saved_tensors
        (
            ln_w,
            ln_b,
            cfc_w,
            cfc_b,
            cproj_w,
            cproj_b,
        )=ctx.param_refs

        block=model.transformer.h[
            ctx.layer_idx
        ]
        G=grad_output.float()

        # Formal D2 winner: nonlinear MLP recomputation consumes the full
        # two-bit fine Z view.
        Zconsumer=unpack_saved_z_fine(
            packed,
            scale,
            p2,
            ctx.mode,
            ctx.z_shape,
        )

        mlp_params=[
            cfc_w,
            cfc_b,
            cproj_w,
            cproj_b,
        ]
        old=set_param_flags(
            mlp_params,
            True,
        )

        try:
            with torch.enable_grad():
                Zreq=(
                    Zconsumer.detach()
                    .to(block.ln_2.weight.dtype)
                    .requires_grad_(True)
                )
                Mhat=block.mlp(
                    Zreq
                )
                surrogate=(
                    Mhat.float()*
                    G
                ).sum()

                (
                    dZ,
                    gcfc_w,
                    gcfc_b,
                    gcproj_w,
                    gcproj_b,
                )=torch.autograd.grad(
                    surrogate,
                    [
                        Zreq,
                        cfc_w,
                        cfc_b,
                        cproj_w,
                        cproj_b,
                    ],
                    retain_graph=False,
                    create_graph=False,
                    allow_unused=False,
                )
        finally:
            restore_param_flags(
                mlp_params,
                old,
            )

        del Zconsumer,Zreq,Mhat,surrogate

        # LN2 remains on the fine normalized-direction sufficient-statistic view.
        Zbase=unpack_saved_z(
            packed,
            scale,
            ctx.mode,
            ctx.z_shape,
        )
        xhat_ln=ln2_xhat_from_saved_dual2(
            Zbase,
            scale,
            p2,
            ctx.mode,
            ctx.layer_idx,
        )

        dU,gln_w,gln_b=explicit_ln2_backward_from_xhat(
            dZ.detach().float(),
            xhat_ln,
            rstd,
            ctx.layer_idx,
            G,
        )
        del Zbase,xhat_ln

        return (
            dU.to(ctx.u_dtype),
            gln_w.to(ln_w.dtype),
            gln_b.to(ln_b.dtype),
            gcfc_w.to(cfc_w.dtype),
            gcfc_b.to(cfc_b.dtype),
            gcproj_w.to(cproj_w.dtype),
            gcproj_b.to(cproj_b.dtype),
            None,
            None,
        )
```

```python
def attention_params_full(layer):
    block=model.transformer.h[layer]
    return [
        block.ln_1.weight,block.ln_1.bias,
        block.attn.c_attn.weight,block.attn.c_attn.bias,
        block.attn.c_proj.weight,block.attn.c_proj.bias,
    ]

def mlp_params_full(layer):
    block=model.transformer.h[layer]
    return [
        block.ln_2.weight,block.ln_2.bias,
        block.mlp.c_fc.weight,block.mlp.c_fc.bias,
        block.mlp.c_proj.weight,block.mlp.c_proj.bias,
    ]

def native_transformer_body(H0):
    H=H0
    for layer in range(12):
        block=model.transformer.h[layer]
        H=H+block.attn(block.ln_1(H),use_cache=False)[0]
        H=H+block.mlp(block.ln_2(H))
    return H

def shv_transformer_body(H0):
    H=H0
    for layer in range(12):
        H=SHVLN1AttentionFineViewFunction.apply(
            H,*attention_params_full(layer),layer,FROZEN_ATTN_X_MODES[layer]
        )
        H=SHVLN2MLPFineViewFunction.apply(
            H,*mlp_params_full(layer),layer,FROZEN_MLP_Z_MODES[layer]
        )
    return H

def unique_named_parameters():
    return list(model.named_parameters(recurse=True,remove_duplicate=True))

def all_unique_parameters():
    return [p for _,p in unique_named_parameters()]

def set_all_model_param_grad_flags(flag):
    for p in model.parameters():
        p.requires_grad_(flag)

def training_embedding_forward(ids):
    B,T=ids.shape
    assert T==SEQ_LEN
    pos=torch.arange(T,device=ids.device,dtype=torch.long).unsqueeze(0).expand(B,T)
    return (model.transformer.wte(ids)+model.transformer.wpe(pos)).to(
        model.transformer.h[0].ln_1.weight.dtype
    )

def native_full_lm_logits(ids):
    H=training_embedding_forward(ids)
    H=native_transformer_body(H)
    H=model.transformer.ln_f(H)
    return model.lm_head(H)

def shv_full_lm_logits(ids):
    H=training_embedding_forward(ids)
    H=shv_transformer_body(H)
    H=model.transformer.ln_f(H)
    return model.lm_head(H)

def causal_lm_loss(logits,ids):
    shift_logits=logits[:,:-1,:].contiguous().float()
    shift_labels=ids[:,1:].contiguous()
    return F.cross_entropy(
        shift_logits.view(-1,shift_logits.shape[-1]),
        shift_labels.view(-1),
    )

def full_lm_loss(ids,kind):
    assert kind in ["native","shv"]
    logits=native_full_lm_logits(ids) if kind=="native" else shv_full_lm_logits(ids)
    return causal_lm_loss(logits,ids)

@torch.no_grad()
def snapshot_state_cpu():
    return {k:v.detach().cpu().clone() for k,v in model.state_dict().items()}

@torch.no_grad()
def restore_state_cpu(state):
    model.load_state_dict(state,strict=True)

@torch.no_grad()
def snapshot_unique_params_cpu():
    return {name:p.detach().cpu().clone() for name,p in unique_named_parameters()}

def state_exact_identity(state):
    worst=0.0
    with torch.no_grad():
        for k,v in model.state_dict().items():
            ref=state[k].to(device=v.device,dtype=v.dtype)
            worst=max(worst,float((v.detach()-ref).abs().max()))
    return worst

def model_parameters_are_finite():
    return all(torch.isfinite(p).all().item() for p in model.parameters())

def build_training_batches():
    batches=[]
    for step in range(TRAIN_STEPS):
        li=step*TRAIN_LANGUAGE_PER_STEP
        ci=step*TRAIN_CODE_PER_STEP
        seqs=[
            PANELS["support"]["language"][li],
            PANELS["support"]["code"][ci],
            PANELS["support"]["language"][li+1],
            PANELS["support"]["code"][ci+1],
        ]
        batch=torch.stack(seqs,dim=0).to(DEVICE,dtype=torch.long)
        assert batch.shape==(TRAIN_BATCH_SIZE,SEQ_LEN)
        batches.append(batch)
    return batches

TRAIN_BATCHES=build_training_batches()
FIXED_PROBE_BATCH=TRAIN_BATCHES[0]
INITIAL_STATE_CPU=snapshot_state_cpu()

class ChunkedMFP16VBF16AdamW:
    # Formal low-memory optimizer; no FP32 master parameter.
    def __init__(self,params,chunk_elems=FORMAL_OPTIMIZER_CHUNK_ELEMS):
        self.params=list(params)
        self.chunk_elems=int(chunk_elems)
        self.step_count=0
        self.exp_avg=[torch.zeros_like(p,dtype=torch.float16) for p in self.params]
        self.exp_avg_sq=[torch.zeros_like(p,dtype=torch.bfloat16) for p in self.params]

    def zero_grad(self,set_to_none=True):
        for p in self.params:
            if set_to_none:
                p.grad=None
            elif p.grad is not None:
                p.grad.zero_()

    @torch.no_grad()
    def step(self):
        self.step_count+=1
        beta1,beta2=ADAMW_BETAS
        bc1=1.0-beta1**self.step_count
        bc2=1.0-beta2**self.step_count
        step_size=ADAMW_LR/bc1
        sqrt_bc2=math.sqrt(bc2)

        for i,p in enumerate(self.params):
            if p.grad is None:
                continue
            if not p.is_contiguous() or not p.grad.is_contiguous():
                raise RuntimeError({"noncontiguous_optimizer_tensor":list(p.shape)})

            pf=p.view(-1)
            gf=p.grad.detach().view(-1)
            mf=self.exp_avg[i].view(-1)
            vf=self.exp_avg_sq[i].view(-1)

            for start in range(0,pf.numel(),self.chunk_elems):
                end=min(start+self.chunk_elems,pf.numel())
                g=gf[start:end].float()
                m_new=(mf[start:end].float()*beta1).add_(g,alpha=1.0-beta1)
                v_new=(vf[start:end].float()*beta2).addcmul_(
                    g,g,value=1.0-beta2
                )
                mf[start:end].copy_(m_new.to(torch.float16))
                vf[start:end].copy_(v_new.to(torch.bfloat16))

                m_use=mf[start:end].float()
                v_use=vf[start:end].float()
                denom=(v_use.sqrt()/sqrt_bc2).add_(ADAMW_EPS)
                target=pf[start:end].float()
                if ADAMW_WEIGHT_DECAY!=0.0:
                    target.mul_(1.0-ADAMW_LR*ADAMW_WEIGHT_DECAY)
                target.addcdiv_(m_use,denom,value=-step_size)
                pf[start:end].copy_(target.to(pf.dtype))

    def state_tensors(self):
        out=[]
        for m,v in zip(self.exp_avg,self.exp_avg_sq):
            out.extend([m,v])
        return out

class ChunkedFP32ReferenceAdamW:
    # Same chunk schedule, FP32 moments; used only as Native optimizer control.
    def __init__(self,params,chunk_elems=FORMAL_OPTIMIZER_CHUNK_ELEMS):
        self.params=list(params)
        self.chunk_elems=int(chunk_elems)
        self.step_count=0
        self.exp_avg=[torch.zeros_like(p,dtype=torch.float32) for p in self.params]
        self.exp_avg_sq=[torch.zeros_like(p,dtype=torch.float32) for p in self.params]

    def zero_grad(self,set_to_none=True):
        for p in self.params:
            if set_to_none:
                p.grad=None
            elif p.grad is not None:
                p.grad.zero_()

    @torch.no_grad()
    def step(self):
        self.step_count+=1
        beta1,beta2=ADAMW_BETAS
        bc1=1.0-beta1**self.step_count
        bc2=1.0-beta2**self.step_count
        step_size=ADAMW_LR/bc1
        sqrt_bc2=math.sqrt(bc2)

        for i,p in enumerate(self.params):
            if p.grad is None:
                continue
            pf=p.view(-1)
            gf=p.grad.detach().view(-1)
            mf=self.exp_avg[i].view(-1)
            vf=self.exp_avg_sq[i].view(-1)
            for start in range(0,pf.numel(),self.chunk_elems):
                end=min(start+self.chunk_elems,pf.numel())
                g=gf[start:end].float()
                m=mf[start:end]
                v=vf[start:end]
                m.mul_(beta1).add_(g,alpha=1.0-beta1)
                v.mul_(beta2).addcmul_(g,g,value=1.0-beta2)
                denom=(v.sqrt()/sqrt_bc2).add_(ADAMW_EPS)
                target=pf[start:end].float()
                if ADAMW_WEIGHT_DECAY!=0.0:
                    target.mul_(1.0-ADAMW_LR*ADAMW_WEIGHT_DECAY)
                target.addcdiv_(m,denom,value=-step_size)
                pf[start:end].copy_(target.to(pf.dtype))

    def state_tensors(self):
        out=[]
        for m,v in zip(self.exp_avg,self.exp_avg_sq):
            out.extend([m,v])
        return out

def make_formal_optimizer():
    assert FORMAL_OPTIMIZER_VARIANT=="mfp16_vbf16_chunk1m"
    return ChunkedMFP16VBF16AdamW(
        all_unique_parameters(),FORMAL_OPTIMIZER_CHUNK_ELEMS
    )

def make_fp32_reference_optimizer():
    return ChunkedFP32ReferenceAdamW(
        all_unique_parameters(),FORMAL_OPTIMIZER_CHUNK_ELEMS
    )

def optimizer_state_byte_summary(opt):
    tensors=opt.state_tensors()
    total=0
    by_dtype={}
    for t in tensors:
        b=int(t.numel()*t.element_size())
        total+=b
        key=str(t.dtype)
        by_dtype[key]=by_dtype.get(key,0)+b
    return {"tensor_count":len(tensors),"total_bytes":int(total),"bytes_by_dtype":by_dtype}

def optimizer_state_is_finite(opt):
    return all(torch.isfinite(t).all().item() for t in opt.state_tensors())

def second_moment_underflow_audit(opt):
    active=0
    zero_second=0
    min_positive=None
    for p,v in zip(all_unique_parameters(),opt.exp_avg_sq):
        if p.grad is None:
            continue
        mask=(p.grad.detach()!=0)
        active_here=int(mask.sum())
        active+=active_here
        if active_here:
            zero_second+=int(((v==0)&mask).sum())
        positive=v[v>0]
        if positive.numel():
            local=float(positive.min())
            min_positive=local if min_positive is None else min(min_positive,local)
    return {
        "active_gradient_elements":int(active),
        "active_elements_with_zero_second_moment":int(zero_second),
        "active_zero_second_moment_fraction":float(zero_second/max(active,1)),
        "minimum_positive_second_moment":min_positive,
    }

@torch.no_grad()
def independent_reference_mfp16_vbf16(initial,gradients):
    p=initial.clone()
    m=torch.zeros_like(initial,dtype=torch.float16)
    v=torch.zeros_like(initial,dtype=torch.bfloat16)
    beta1,beta2=ADAMW_BETAS
    for step,g16 in enumerate(gradients,start=1):
        g=g16.float()
        m=((m.float()*beta1).add(g,alpha=1.0-beta1)).to(torch.float16)
        v=((v.float()*beta2).addcmul(g,g,value=1.0-beta2)).to(torch.bfloat16)
        bc1=1.0-beta1**step
        bc2=1.0-beta2**step
        denom=v.float().sqrt()/math.sqrt(bc2)+ADAMW_EPS
        target=p.float()
        if ADAMW_WEIGHT_DECAY!=0.0:
            target=target*(1.0-ADAMW_LR*ADAMW_WEIGHT_DECAY)
        target=target-(ADAMW_LR/bc1)*m.float()/denom
        p=target.to(torch.float16)
    return p,m,v

class FormulaAuditMFP16VBF16AdamW(ChunkedMFP16VBF16AdamW):
    pass

@torch.no_grad()
def optimizer_formula_audit():
    initial=torch.linspace(-1.0,1.0,4099,device=DEVICE,dtype=torch.float16)
    gen=torch.Generator(device=DEVICE)
    gen.manual_seed(SEED+12051)
    gradients=[
        (torch.randn(initial.shape,generator=gen,device=DEVICE,dtype=torch.float32)*scale).to(torch.float16)
        for scale in [1e-2,1e-3,2e-4]
    ]
    ref_p,ref_m,ref_v=independent_reference_mfp16_vbf16(initial,gradients)
    p=torch.nn.Parameter(initial.clone())
    opt=FormulaAuditMFP16VBF16AdamW([p],chunk_elems=257)
    for g in gradients:
        p.grad=g.clone()
        opt.step()
    result={
        "weight_exact":bool(torch.equal(p.detach(),ref_p)),
        "first_moment_exact":bool(torch.equal(opt.exp_avg[0],ref_m)),
        "second_moment_exact":bool(torch.equal(opt.exp_avg_sq[0],ref_v)),
        "weight_maxabs":float((p.detach().float()-ref_p.float()).abs().max()),
    }
    result["passed"]=bool(
        result["weight_exact"] and result["first_moment_exact"] and result["second_moment_exact"]
    )
    return result

OPTIMIZER_FORMULA_AUDIT=optimizer_formula_audit()
(OUT/"optimizer_formula_audit.json").write_text(
    json.dumps(OPTIMIZER_FORMULA_AUDIT,indent=2),encoding="utf-8"
)
print(json.dumps(OPTIMIZER_FORMULA_AUDIT,indent=2))
if not OPTIMIZER_FORMULA_AUDIT["passed"]:
    raise RuntimeError({"optimizer_formula_audit_failed":OPTIMIZER_FORMULA_AUDIT})

probe_ids=TRAIN_BATCHES[0]
with torch.no_grad():
    hf_logits=model(input_ids=probe_ids,use_cache=False).logits
    manual_logits=native_full_lm_logits(probe_ids)
    shv_logits=shv_full_lm_logits(probe_ids)
    manual_vs_hf=nrmse(hf_logits,manual_logits)
    native_vs_shv=nrmse(manual_logits,shv_logits)
    loss_absdiff=float(
        (causal_lm_loss(manual_logits,probe_ids)-causal_lm_loss(shv_logits,probe_ids)).abs()
    )
del hf_logits,manual_logits,shv_logits
cleanup()

restore_state_cpu(INITIAL_STATE_CPU)
set_all_model_param_grad_flags(True)
census_opt=make_formal_optimizer()
OPTIMIZER_STATE_CENSUS=optimizer_state_byte_summary(census_opt)
del census_opt
cleanup()

OPTIMIZER_STATE_CENSUS["passed"]=bool(
    OPTIMIZER_STATE_CENSUS["tensor_count"]==296
    and OPTIMIZER_STATE_CENSUS["total_bytes"]==EXPECTED_LOW_MEMORY_MOMENT_BYTES
    and OPTIMIZER_STATE_CENSUS["bytes_by_dtype"]=={
        EXPECTED_FIRST_MOMENT_DTYPE:EXPECTED_FIRST_MOMENT_BYTES,
        EXPECTED_SECOND_MOMENT_DTYPE:EXPECTED_SECOND_MOMENT_BYTES,
    }
)
(OUT/"optimizer_state_census.json").write_text(
    json.dumps(OPTIMIZER_STATE_CENSUS,indent=2),encoding="utf-8"
)

PRECHECK={
    "fresh_hash_audit_pass":bool(FINAL_HASH_AUDIT["all_passed"]),
    "support_language_count":len(PANELS["support"]["language"]),
    "support_code_count":len(PANELS["support"]["code"]),
    "promotion_language_count":len(PANELS["promotion"]["language"]),
    "promotion_code_count":len(PANELS["promotion"]["code"]),
    "guard_language_count":len(PANELS["guard"]["language"]),
    "guard_code_count":len(PANELS["guard"]["code"]),
    "manual_native_vs_hf_logits_nrmse":float(manual_vs_hf),
    "initial_native_vs_shv_forward_nrmse":float(native_vs_shv),
    "initial_native_vs_shv_loss_absdiff":float(loss_absdiff),
    "optimizer_formula_audit_pass":bool(OPTIMIZER_FORMULA_AUDIT["passed"]),
    "optimizer_state_census_pass":bool(OPTIMIZER_STATE_CENSUS["passed"]),
    "all_model_parameters_contiguous":bool(all(p.is_contiguous() for p in all_unique_parameters())),
    "initial_state_identity_maxabs":float(state_exact_identity(INITIAL_STATE_CPU)),
}
PRECHECK["passed"]=bool(
    PRECHECK["fresh_hash_audit_pass"]
    and PRECHECK["support_language_count"]==SUPPORT_N
    and PRECHECK["support_code_count"]==SUPPORT_N
    and PRECHECK["promotion_language_count"]==PROMOTION_N
    and PRECHECK["promotion_code_count"]==PROMOTION_N
    and PRECHECK["guard_language_count"]==GUARD_N
    and PRECHECK["guard_code_count"]==GUARD_N
    and PRECHECK["manual_native_vs_hf_logits_nrmse"]<=FORWARD_IDENTITY_NRMSE_MAX
    and PRECHECK["initial_native_vs_shv_forward_nrmse"]<=FORWARD_IDENTITY_NRMSE_MAX
    and PRECHECK["initial_native_vs_shv_loss_absdiff"]<=1e-7
    and PRECHECK["optimizer_formula_audit_pass"]
    and PRECHECK["optimizer_state_census_pass"]
    and PRECHECK["all_model_parameters_contiguous"]
    and PRECHECK["initial_state_identity_maxabs"]==0.0
)
(OUT/"preflight_qa.json").write_text(json.dumps(PRECHECK,indent=2),encoding="utf-8")
print(json.dumps(PRECHECK,indent=2))
if not PRECHECK["passed"]:
    raise RuntimeError({"R5B0R2_preflight_failed":PRECHECK})
```

```python
INITIAL_UNIQUE={name:INITIAL_STATE_CPU[name] for name,_ in unique_named_parameters()}

class HeterocodedMomentAdamW:
    # Persistent state:
    #   moment_pair_carrier[:,0] = q_v
    #   moment_pair_carrier[:,1] = q_r + 128
    # plus FP16 per-group sidecars.

    def __init__(
        self,
        params,
        chunk_elems=MOMENT_HETEROCODE_CHUNK_ELEMS,
        group_size=MOMENT_HETEROCODE_GROUP_SIZE,
    ):
        self.params=list(params)
        self.chunk_elems=int(chunk_elems)
        self.group_size=int(group_size)
        assert self.chunk_elems % self.group_size == 0
        self.step_count=0

        self.moment_pair_carrier=[
            torch.zeros((p.numel(),2),device=p.device,dtype=torch.uint8)
            for p in self.params
        ]
        self.v_log_min=[]
        self.v_log_step=[]
        self.r_scale=[]

        for p in self.params:
            ng=(p.numel()+self.group_size-1)//self.group_size
            self.v_log_min.append(torch.zeros(ng,device=p.device,dtype=torch.float16))
            self.v_log_step.append(torch.zeros(ng,device=p.device,dtype=torch.float16))
            self.r_scale.append(torch.zeros(ng,device=p.device,dtype=torch.float16))

        self.last_positive_v_count=0
        self.last_positive_v_to_zero_count=0

    def zero_grad(self,set_to_none=True):
        for p in self.params:
            if set_to_none:
                p.grad=None
            elif p.grad is not None:
                p.grad.zero_()

    def _decode_chunk(self,i,start,end):
        G=self.group_size
        L=end-start
        assert start % G == 0

        ng=(L+G-1)//G
        P=ng*G
        g0=start//G

        pair=self.moment_pair_carrier[i][start:end]
        qv_flat=torch.zeros(P,device=pair.device,dtype=torch.int16)
        qr_flat=torch.zeros(P,device=pair.device,dtype=torch.int16)

        qv_flat[:L]=pair[:,0].to(torch.int16)
        qr_flat[:L]=pair[:,1].to(torch.int16)-128

        qv=qv_flat.view(ng,G)
        qr=qr_flat.view(ng,G)

        lo=self.v_log_min[i][g0:g0+ng].float().unsqueeze(1)
        step=self.v_log_step[i][g0:g0+ng].float().unsqueeze(1)
        rs=self.r_scale[i][g0:g0+ng].float().unsqueeze(1)

        positive=qv>0
        lv=lo+(qv.float()-1.0)*step
        v=torch.where(positive,torch.exp2(lv),torch.zeros_like(lv))

        r=qr.float()*rs
        m=r*(v.sqrt()+ADAMW_EPS)

        return m.reshape(-1)[:L],v.reshape(-1)[:L]

    @torch.no_grad()
    def _encode_chunk(self,i,start,end,m_new,v_new):
        G=self.group_size
        L=end-start
        assert start % G == 0

        ng=(L+G-1)//G
        P=ng*G
        g0=start//G

        m_pad=torch.zeros(P,device=m_new.device,dtype=torch.float32)
        v_pad=torch.zeros(P,device=v_new.device,dtype=torch.float32)
        m_pad[:L]=m_new
        v_pad[:L]=v_new

        m2=m_pad.view(ng,G)
        v2=v_pad.view(ng,G)

        positive=v2>0
        lv=torch.log2(v2.clamp_min(1e-30))

        lo_raw=torch.where(
            positive,lv,torch.full_like(lv,float("inf"))
        ).amin(dim=1)
        hi_raw=torch.where(
            positive,lv,torch.full_like(lv,float("-inf"))
        ).amax(dim=1)
        has=positive.any(dim=1)

        lo=torch.where(has,lo_raw,torch.zeros_like(lo_raw))
        hi=torch.where(has,hi_raw,lo)
        span=(hi-lo).clamp_min(0.0)
        step=torch.where(span>0,span/254.0,torch.zeros_like(span))
        step_safe=torch.where(step>0,step,torch.ones_like(step))

        qv_pos=1+torch.round(
            (lv-lo.unsqueeze(1))/step_safe.unsqueeze(1)
        ).clamp(0,254)
        qv=torch.where(
            positive,qv_pos,torch.zeros_like(qv_pos)
        ).to(torch.uint8)

        ratio=m2/(v2.sqrt()+ADAMW_EPS)
        max_abs=ratio.abs().amax(dim=1)
        rs=torch.where(max_abs>0,max_abs/127.0,torch.zeros_like(max_abs))
        rs_safe=torch.where(rs>0,rs,torch.ones_like(rs))

        qr=torch.round(
            ratio/rs_safe.unsqueeze(1)
        ).clamp(-127,127)
        qr=torch.where(
            rs.unsqueeze(1)>0,qr,torch.zeros_like(qr)
        ).to(torch.int16)
        qu=(qr+128).to(torch.uint8)

        # Persist quantized sidecars and carrier.
        self.v_log_min[i][g0:g0+ng].copy_(lo.to(torch.float16))
        self.v_log_step[i][g0:g0+ng].copy_(step.to(torch.float16))
        self.r_scale[i][g0:g0+ng].copy_(rs.to(torch.float16))

        pair=self.moment_pair_carrier[i][start:end]
        pair[:,0].copy_(qv.reshape(-1)[:L])
        pair[:,1].copy_(qu.reshape(-1)[:L])

    @torch.no_grad()
    def step(self):
        self.step_count+=1
        beta1,beta2=ADAMW_BETAS
        bc1=1.0-beta1**self.step_count
        bc2=1.0-beta2**self.step_count
        step_size=ADAMW_LR/bc1
        sqrt_bc2=math.sqrt(bc2)

        positive_v_count=0
        positive_v_to_zero=0

        for i,p in enumerate(self.params):
            if p.grad is None:
                continue
            if not p.is_contiguous() or not p.grad.is_contiguous():
                raise RuntimeError("optimizer requires contiguous tensors")

            pf=p.view(-1)
            gf=p.grad.detach().view(-1)

            for start in range(0,pf.numel(),self.chunk_elems):
                end=min(start+self.chunk_elems,pf.numel())

                old_m,old_v=self._decode_chunk(i,start,end)
                g=gf[start:end].float()

                m_new=(old_m*beta1).add_(g,alpha=1.0-beta1)
                v_new=(old_v*beta2).addcmul_(g,g,value=1.0-beta2)

                self._encode_chunk(i,start,end,m_new,v_new)
                m_use,v_use=self._decode_chunk(i,start,end)

                pos=v_new>0
                positive_v_count+=int(pos.sum())
                positive_v_to_zero+=int((pos & (v_use==0)).sum())

                denom=(v_use.sqrt()/sqrt_bc2).add_(ADAMW_EPS)
                target=pf[start:end].float()

                if ADAMW_WEIGHT_DECAY!=0.0:
                    target.mul_(1.0-ADAMW_LR*ADAMW_WEIGHT_DECAY)

                target.addcdiv_(m_use,denom,value=-step_size)
                pf[start:end].copy_(target.to(pf.dtype))

        self.last_positive_v_count=int(positive_v_count)
        self.last_positive_v_to_zero_count=int(positive_v_to_zero)

    def state_tensors(self):
        out=[]
        for i in range(len(self.params)):
            out.extend([
                self.moment_pair_carrier[i],
                self.v_log_min[i],
                self.v_log_step[i],
                self.r_scale[i],
            ])
        return out

def make_heterocoded_optimizer():
    return HeterocodedMomentAdamW(
        all_unique_parameters(),
        MOMENT_HETEROCODE_CHUNK_ELEMS,
        MOMENT_HETEROCODE_GROUP_SIZE,
    )

def make_conventional_optimizer():
    return make_formal_optimizer()

def state_bytes(opt):
    return optimizer_state_byte_summary(opt)

restore_state_cpu(INITIAL_STATE_CPU)
conv=make_conventional_optimizer()
het=make_heterocoded_optimizer()

CONVENTIONAL_STATE_CENSUS=state_bytes(conv)
HETERO_STATE_CENSUS=state_bytes(het)
HETERO_STATE_CENSUS["ratio_vs_conventional"]=float(
    HETERO_STATE_CENSUS["total_bytes"]/
    CONVENTIONAL_STATE_CENSUS["total_bytes"]
)

STATE_CENSUS={
    "conventional":CONVENTIONAL_STATE_CENSUS,
    "optimizer_state_heterocode":HETERO_STATE_CENSUS,
    "passed":bool(
        CONVENTIONAL_STATE_CENSUS["total_bytes"]==
        CURRENT_QUALIFIED_OPTIMIZER_STATE_BYTES
        and
        HETERO_STATE_CENSUS["ratio_vs_conventional"]<=
        HETERO_STATE_BYTE_RATIO_MAX
    ),
}

del conv,het
cleanup()

(OUT/"optimizer_state_carrier_census.json").write_text(
    json.dumps(STATE_CENSUS,indent=2),encoding="utf-8"
)
print(json.dumps(STATE_CENSUS,indent=2))

if not STATE_CENSUS["passed"]:
    raise RuntimeError({"optimizer_state_carrier_census_failed":STATE_CENSUS})


@torch.no_grad()
def reference_encode_moment_pair(m_new,v_new,group_size):
    G=int(group_size)
    L=int(m_new.numel())
    ng=(L+G-1)//G
    P=ng*G

    m_pad=torch.zeros(P,device=m_new.device,dtype=torch.float32)
    v_pad=torch.zeros(P,device=v_new.device,dtype=torch.float32)
    m_pad[:L]=m_new.float()
    v_pad[:L]=v_new.float()

    m2=m_pad.view(ng,G)
    v2=v_pad.view(ng,G)

    positive=v2>0
    lv=torch.log2(v2.clamp_min(1e-30))

    lo_raw=torch.where(
        positive,lv,torch.full_like(lv,float("inf"))
    ).amin(dim=1)
    hi_raw=torch.where(
        positive,lv,torch.full_like(lv,float("-inf"))
    ).amax(dim=1)

    has=positive.any(dim=1)
    lo=torch.where(has,lo_raw,torch.zeros_like(lo_raw))
    hi=torch.where(has,hi_raw,lo)

    span=(hi-lo).clamp_min(0.0)
    step=torch.where(span>0,span/254.0,torch.zeros_like(span))
    step_safe=torch.where(step>0,step,torch.ones_like(step))

    qv_pos=1+torch.round(
        (lv-lo.unsqueeze(1))/step_safe.unsqueeze(1)
    ).clamp(0,254)

    qv=torch.where(
        positive,qv_pos,torch.zeros_like(qv_pos)
    ).to(torch.uint8)

    ratio=m2/(v2.sqrt()+ADAMW_EPS)
    max_abs=ratio.abs().amax(dim=1)

    rs=torch.where(
        max_abs>0,max_abs/127.0,torch.zeros_like(max_abs)
    )
    rs_safe=torch.where(rs>0,rs,torch.ones_like(rs))

    qr=torch.round(
        ratio/rs_safe.unsqueeze(1)
    ).clamp(-127,127)
    qr=torch.where(
        rs.unsqueeze(1)>0,qr,torch.zeros_like(qr)
    ).to(torch.int16)

    pair=torch.stack(
        [
            qv.reshape(-1)[:L],
            (qr+128).to(torch.uint8).reshape(-1)[:L],
        ],
        dim=1,
    )

    return (
        pair,
        lo.to(torch.float16),
        step.to(torch.float16),
        rs.to(torch.float16),
    )


@torch.no_grad()
def reference_decode_moment_pair(
    pair,log_min,log_step,ratio_scale,group_size
):
    G=int(group_size)
    L=int(pair.shape[0])
    ng=(L+G-1)//G
    P=ng*G

    qv_flat=torch.zeros(P,device=pair.device,dtype=torch.int16)
    qr_flat=torch.zeros(P,device=pair.device,dtype=torch.int16)

    qv_flat[:L]=pair[:,0].to(torch.int16)
    qr_flat[:L]=pair[:,1].to(torch.int16)-128

    qv=qv_flat.view(ng,G)
    qr=qr_flat.view(ng,G)

    lo=log_min.float().unsqueeze(1)
    step=log_step.float().unsqueeze(1)
    rs=ratio_scale.float().unsqueeze(1)

    positive=qv>0
    lv=lo+(qv.float()-1.0)*step

    v=torch.where(
        positive,torch.exp2(lv),torch.zeros_like(lv)
    )
    r=qr.float()*rs
    m=r*(v.sqrt()+ADAMW_EPS)

    return (
        m.reshape(-1)[:L],
        v.reshape(-1)[:L],
    )


@torch.no_grad()
def heterocode_reference_audit():
    gen=torch.Generator(device=DEVICE)
    gen.manual_seed(SEED+15071)

    L=1031
    g=torch.randn(
        L,generator=gen,device=DEVICE,dtype=torch.float32
    )*1e-3
    old_m=torch.randn(
        L,generator=gen,device=DEVICE,dtype=torch.float32
    )*5e-4
    old_v=torch.rand(
        L,generator=gen,device=DEVICE,dtype=torch.float32
    )*1e-5
    old_v[::37]=0.0

    beta1,beta2=ADAMW_BETAS
    m_new=(old_m*beta1).add(g,alpha=1.0-beta1)
    v_new=(old_v*beta2).addcmul(g,g,value=1.0-beta2)

    ref_pair,ref_lo,ref_step,ref_rs=reference_encode_moment_pair(
        m_new,v_new,R5C1_FORMAL_GROUP_SIZE
    )

    p=torch.nn.Parameter(
        torch.zeros(L,device=DEVICE,dtype=torch.float16)
    )

    opt=HeterocodedMomentAdamW(
        [p],
        chunk_elems=R5C1_FORMAL_GROUP_SIZE*8,
        group_size=R5C1_FORMAL_GROUP_SIZE,
    )

    opt._encode_chunk(0,0,L,m_new,v_new)

    prod_pair=opt.moment_pair_carrier[0].clone()
    prod_lo=opt.v_log_min[0].clone()
    prod_step=opt.v_log_step[0].clone()
    prod_rs=opt.r_scale[0].clone()

    prod_m,prod_v=opt._decode_chunk(0,0,L)
    ref_m,ref_v=reference_decode_moment_pair(
        ref_pair,ref_lo,ref_step,ref_rs,R5C1_FORMAL_GROUP_SIZE
    )

    result={
        "carrier_exact":bool(torch.equal(prod_pair,ref_pair)),
        "log_min_exact":bool(torch.equal(prod_lo,ref_lo)),
        "log_step_exact":bool(torch.equal(prod_step,ref_step)),
        "ratio_scale_exact":bool(torch.equal(prod_rs,ref_rs)),
        "decoded_m_exact":bool(torch.equal(prod_m,ref_m)),
        "decoded_v_exact":bool(torch.equal(prod_v,ref_v)),
        "positive_v_decoded_as_zero":int(
            ((v_new>0)&(prod_v==0)).sum()
        ),
    }

    result["passed"]=bool(
        result["carrier_exact"]
        and result["log_min_exact"]
        and result["log_step_exact"]
        and result["ratio_scale_exact"]
        and result["decoded_m_exact"]
        and result["decoded_v_exact"]
        and result["positive_v_decoded_as_zero"]==0
    )
    return result


HETEROCODE_REFERENCE_AUDIT=heterocode_reference_audit()

(OUT/"heterocode_reference_audit.json").write_text(
    json.dumps(HETEROCODE_REFERENCE_AUDIT,indent=2),
    encoding="utf-8",
)

R5C1_STATE_CENSUS_PASS=bool(
    STATE_CENSUS["passed"]
    and
    HETERO_STATE_CENSUS["total_bytes"]==
    R5C1_EXPECTED_HETERO_STATE_BYTES
    and
    HETERO_STATE_CENSUS["ratio_vs_conventional"]<=
    R5C1_EXPECTED_STATE_RATIO_MAX
)

R5C1_PRECHECK={
    "parent_precheck_pass":bool(PRECHECK["passed"]),
    "fresh_hash_audit_pass":bool(FINAL_HASH_AUDIT["all_passed"]),
    "heterocode_reference_audit":HETEROCODE_REFERENCE_AUDIT,
    "state_census_pass":R5C1_STATE_CENSUS_PASS,
    "heterocoded_state_bytes":HETERO_STATE_CENSUS["total_bytes"],
    "heterocoded_state_ratio":HETERO_STATE_CENSUS[
        "ratio_vs_conventional"
    ],
}

R5C1_PRECHECK["passed"]=bool(
    R5C1_PRECHECK["parent_precheck_pass"]
    and R5C1_PRECHECK["fresh_hash_audit_pass"]
    and HETEROCODE_REFERENCE_AUDIT["passed"]
    and R5C1_STATE_CENSUS_PASS
)

(OUT/"r5c1_preflight_qa.json").write_text(
    json.dumps(R5C1_PRECHECK,indent=2),
    encoding="utf-8",
)

print(json.dumps(HETEROCODE_REFERENCE_AUDIT,indent=2))
print(json.dumps(R5C1_PRECHECK,indent=2))

if not R5C1_PRECHECK["passed"]:
    raise RuntimeError({"R5C1_preflight_failed":R5C1_PRECHECK})
```

```python
def safe_tensor_cosine(a,b):
    a=a.float().reshape(-1)
    b=b.float().reshape(-1)
    na=float(a.norm())
    nb=float(b.norm())

    if na<=ACTIVE_GRAD_NORM_EPS and nb<=ACTIVE_GRAD_NORM_EPS:
        return 1.0,False,na,nb
    if na<=ACTIVE_GRAD_NORM_EPS or nb<=ACTIVE_GRAD_NORM_EPS:
        return 0.0,True,na,nb

    return (
        float(torch.dot(a,b)/(a.norm()*b.norm()+1e-30)),
        True,na,nb,
    )


def collect_gradient_snapshot(kind,batch):
    restore_state_cpu(INITIAL_STATE_CPU)
    set_all_model_param_grad_flags(True)
    for p in model.parameters():
        p.grad=None

    loss=full_lm_loss(batch,kind)
    loss.backward()

    snap={
        name:(
            None
            if p.grad is None
            else p.grad.detach().float().cpu().clone()
        )
        for name,p in unique_named_parameters()
    }
    value=float(loss.detach())

    for p in model.parameters():
        p.grad=None
    del loss
    cleanup()
    return value,snap


nloss,ngrads=collect_gradient_snapshot("native",TRAIN_BATCHES[0])
sloss,sgrads=collect_gradient_snapshot("shv",TRAIN_BATCHES[0])

rows=[]
dot=n2=s2=0.0

for name,_ in unique_named_parameters():
    gn=ngrads[name]
    gs=sgrads[name]

    if gn is None and gs is None:
        cos,active,nn,ns=1.0,False,0.0,0.0
    elif gn is None or gs is None:
        cos,active=0.0,True
        nn=0.0 if gn is None else float(gn.norm())
        ns=0.0 if gs is None else float(gs.norm())
    else:
        cos,active,nn,ns=safe_tensor_cosine(gn,gs)
        dot+=float(torch.dot(gn.reshape(-1),gs.reshape(-1)))
        n2+=float(gn.pow(2).sum())
        s2+=float(gs.pow(2).sum())

    rows.append({
        "parameter":name,
        "active":bool(active),
        "native_norm":nn,
        "shv_norm":ns,
        "gradient_cosine":cos,
    })

FIRST_STEP_GRADS=pd.DataFrame(rows)
FIRST_STEP_GRADS.to_csv(
    OUT/"first_step_parameter_gradient_comparison.csv",
    index=False,
)

active=FIRST_STEP_GRADS[FIRST_STEP_GRADS.active]

FIRST_STEP_GRAD_SUMMARY={
    "native_loss":nloss,
    "shv_loss":sloss,
    "loss_absdiff":abs(nloss-sloss),
    "global_gradient_cosine":float(
        dot/(math.sqrt(n2)*math.sqrt(s2)+1e-30)
    ),
    "min_active_parameter_gradient_cosine":float(
        active.gradient_cosine.min()
    ),
    "worst_active_parameter":str(
        active.loc[
            active.gradient_cosine.idxmin(),
            "parameter",
        ]
    ),
}

FIRST_STEP_GRAD_SUMMARY["passed"]=bool(
    FIRST_STEP_GRAD_SUMMARY["loss_absdiff"]<=1e-7
    and FIRST_STEP_GRAD_SUMMARY["global_gradient_cosine"]>=FIRST_STEP_GLOBAL_GRAD_COS_MIN
    and FIRST_STEP_GRAD_SUMMARY["min_active_parameter_gradient_cosine"]>=FIRST_STEP_ACTIVE_PARAM_GRAD_COS_MIN
)

(OUT/"first_step_gradient_summary.json").write_text(
    json.dumps(FIRST_STEP_GRAD_SUMMARY,indent=2),
    encoding="utf-8",
)

del ngrads,sgrads
cleanup()


@torch.no_grad()
def update_metrics(initial,a,b):
    rows=[]
    dot=a2=b2=diff2=0.0

    for name,_ in unique_named_parameters():
        init=initial[name].float()
        da=a[name].float()-init
        db=b[name].float()-init

        na=float(da.norm())
        nb=float(db.norm())

        if na<=UPDATE_NORM_EPS and nb<=UPDATE_NORM_EPS:
            cos=1.0
            material=False
        elif na<=UPDATE_NORM_EPS or nb<=UPDATE_NORM_EPS:
            cos=0.0
            material=True
        else:
            cos=float(
                torch.dot(da.reshape(-1),db.reshape(-1))/
                (da.norm()*db.norm()+1e-30)
            )
            material=True

        dot+=float(torch.dot(da.reshape(-1),db.reshape(-1)))
        a2+=float(da.pow(2).sum())
        b2+=float(db.pow(2).sum())
        diff2+=float((db-da).pow(2).sum())

        rows.append({
            "parameter":name,
            "material_update":bool(material),
            "a_update_norm":na,
            "b_update_norm":nb,
            "update_cosine":cos,
        })

    detail=pd.DataFrame(rows)

    return detail,{
        "global_cosine":float(
            dot/(math.sqrt(a2)*math.sqrt(b2)+1e-30)
        ),
        "relative_error":float(
            math.sqrt(diff2)/(math.sqrt(a2)+1e-30)
        ),
        "min_material_parameter_cosine":float(
            detail.loc[
                detail.material_update,
                "update_cosine",
            ].min()
        ),
    }


def run_formal_trajectory(kind,optimizer_kind):
    assert kind in ["native","shv"]
    assert optimizer_kind in ["conventional","heterocoded"]

    restore_state_cpu(INITIAL_STATE_CPU)
    set_all_model_param_grad_flags(True)
    for p in model.parameters():
        p.grad=None

    opt=(
        make_conventional_optimizer()
        if optimizer_kind=="conventional"
        else make_heterocoded_optimizer()
    )

    rows=[]
    first_carrier_audit=None

    for step,batch in enumerate(TRAIN_BATCHES):
        opt.zero_grad(set_to_none=True)

        loss=full_lm_loss(batch,kind)
        if not torch.isfinite(loss).item():
            raise RuntimeError({
                "nonfinite_loss":{
                    "kind":kind,
                    "optimizer_kind":optimizer_kind,
                    "step":step,
                }
            })

        loss.backward()
        grad_norm=torch.nn.utils.clip_grad_norm_(
            all_unique_parameters(),GRAD_CLIP_NORM
        )
        opt.step()

        if step==0 and optimizer_kind=="heterocoded":
            first_carrier_audit={
                "positive_updated_v_count":int(
                    opt.last_positive_v_count
                ),
                "positive_v_decoded_as_zero_count":int(
                    opt.last_positive_v_to_zero_count
                ),
            }

        if not model_parameters_are_finite():
            raise RuntimeError({
                "nonfinite_model":{
                    "kind":kind,
                    "optimizer_kind":optimizer_kind,
                    "step":step,
                }
            })

        if not optimizer_state_is_finite(opt):
            raise RuntimeError({
                "nonfinite_optimizer_state":{
                    "kind":kind,
                    "optimizer_kind":optimizer_kind,
                    "step":step,
                }
            })

        with torch.no_grad():
            probe_logits=native_full_lm_logits(FIXED_PROBE_BATCH)
            probe_loss=causal_lm_loss(
                probe_logits,FIXED_PROBE_BATCH
            )

        rows.append({
            "kind":kind,
            "optimizer_kind":optimizer_kind,
            "step":step,
            "train_loss":float(loss.detach()),
            "preclip_grad_norm":float(grad_norm.detach()),
            "fixed_probe_post_step_loss":float(probe_loss),
        })

        del loss,grad_norm,probe_logits,probe_loss
        cleanup()

    final=snapshot_unique_params_cpu()
    state_summary=optimizer_state_byte_summary(opt)

    for p in model.parameters():
        p.grad=None
    del opt
    cleanup()

    return pd.DataFrame(rows),final,state_summary,first_carrier_audit


print("R5C1 conventional Native")
CONV_NATIVE_TRAJ,CONV_NATIVE_FINAL,CONV_STATE,_=run_formal_trajectory(
    "native","conventional"
)

print("R5C1 heterocoded Native")
HET_NATIVE_TRAJ,HET_NATIVE_FINAL,HET_NATIVE_STATE,HET_NATIVE_AUDIT=run_formal_trajectory(
    "native","heterocoded"
)

print("R5C1 heterocoded SHV")
HET_SHV_TRAJ,HET_SHV_FINAL,HET_SHV_STATE,HET_SHV_AUDIT=run_formal_trajectory(
    "shv","heterocoded"
)

CONV_NATIVE_TRAJ.to_csv(
    OUT/"conventional_native_trajectory.csv",index=False
)
HET_NATIVE_TRAJ.to_csv(
    OUT/"heterocoded_native_trajectory.csv",index=False
)
HET_SHV_TRAJ.to_csv(
    OUT/"heterocoded_shv_trajectory.csv",index=False
)

paired=HET_NATIVE_TRAJ.merge(
    HET_SHV_TRAJ,on="step",suffixes=("_native","_shv")
)
paired["train_loss_relative_gap"]=(
    (paired.train_loss_shv-paired.train_loss_native).abs()/
    paired.train_loss_native.abs().clip(lower=1e-12)
)
paired["fixed_probe_loss_relative_gap"]=(
    (paired.fixed_probe_post_step_loss_shv-paired.fixed_probe_post_step_loss_native).abs()/
    paired.fixed_probe_post_step_loss_native.abs().clip(lower=1e-12)
)
paired.to_csv(
    OUT/"heterocoded_native_shv_paired_trajectory.csv",index=False
)

ns_detail,ns_update=update_metrics(
    INITIAL_UNIQUE,HET_NATIVE_FINAL,HET_SHV_FINAL
)
ns_detail.to_csv(
    OUT/"heterocoded_native_shv_update_detail.csv",index=False
)

HETERO_NATIVE_SHV_SUMMARY={
    "max_train_loss_relative_gap":float(
        paired.train_loss_relative_gap.max()
    ),
    "max_fixed_probe_loss_relative_gap":float(
        paired.fixed_probe_loss_relative_gap.max()
    ),
    "final_update_global_cosine":ns_update["global_cosine"],
    "final_update_relative_error":ns_update["relative_error"],
    "min_material_parameter_update_cosine_record_only":ns_update[
        "min_material_parameter_cosine"
    ],
    "native_first_carrier_audit":HET_NATIVE_AUDIT,
    "shv_first_carrier_audit":HET_SHV_AUDIT,
}

HETERO_NATIVE_SHV_SUMMARY["passed"]=bool(
    FIRST_STEP_GRAD_SUMMARY["passed"]
    and HETERO_NATIVE_SHV_SUMMARY["max_train_loss_relative_gap"]<=R5C1_NATIVE_SHV_LOSS_GAP_MAX
    and HETERO_NATIVE_SHV_SUMMARY["max_fixed_probe_loss_relative_gap"]<=R5C1_NATIVE_SHV_LOSS_GAP_MAX
    and HETERO_NATIVE_SHV_SUMMARY["final_update_global_cosine"]>=R5C1_NATIVE_SHV_UPDATE_COS_MIN
    and HETERO_NATIVE_SHV_SUMMARY["final_update_relative_error"]<=R5C1_NATIVE_SHV_UPDATE_RELERR_MAX
    and HET_NATIVE_AUDIT["positive_v_decoded_as_zero_count"]==0
    and HET_SHV_AUDIT["positive_v_decoded_as_zero_count"]==0
)

ctrl_detail,ctrl_update=update_metrics(
    INITIAL_UNIQUE,CONV_NATIVE_FINAL,HET_NATIVE_FINAL
)
ctrl_detail.to_csv(
    OUT/"heterocoded_native_vs_conventional_native_update_detail.csv",
    index=False,
)

ctrl=CONV_NATIVE_TRAJ[
    ["step","fixed_probe_post_step_loss"]
].rename(
    columns={"fixed_probe_post_step_loss":"conventional_probe_loss"}
).merge(
    HET_NATIVE_TRAJ[
        ["step","fixed_probe_post_step_loss"]
    ].rename(
        columns={"fixed_probe_post_step_loss":"heterocoded_probe_loss"}
    ),
    on="step",
)
ctrl["relative_probe_gap"]=(
    (ctrl.heterocoded_probe_loss-ctrl.conventional_probe_loss).abs()/
    ctrl.conventional_probe_loss.abs().clip(lower=1e-12)
)
ctrl.to_csv(
    OUT/"heterocoded_native_vs_conventional_probe_trajectory.csv",
    index=False,
)

HETERO_NATIVE_CONTROL_SUMMARY={
    "final_update_global_cosine":ctrl_update["global_cosine"],
    "final_update_relative_error":ctrl_update["relative_error"],
    "min_material_parameter_update_cosine_record_only":ctrl_update[
        "min_material_parameter_cosine"
    ],
    "max_fixed_probe_loss_relative_gap":float(
        ctrl.relative_probe_gap.max()
    ),
}
HETERO_NATIVE_CONTROL_SUMMARY["passed"]=bool(
    HETERO_NATIVE_CONTROL_SUMMARY["final_update_global_cosine"]>=R5C1_NATIVE_CONTROL_UPDATE_COS_MIN
    and HETERO_NATIVE_CONTROL_SUMMARY["final_update_relative_error"]<=R5C1_NATIVE_CONTROL_UPDATE_RELERR_MAX
    and HETERO_NATIVE_CONTROL_SUMMARY["max_fixed_probe_loss_relative_gap"]<=R5C1_NATIVE_CONTROL_PROBE_GAP_MAX
)

combined_detail,combined_update=update_metrics(
    INITIAL_UNIQUE,CONV_NATIVE_FINAL,HET_SHV_FINAL
)
combined_detail.to_csv(
    OUT/"combined_conventional_native_vs_heterocoded_shv_update_detail.csv",
    index=False,
)

COMBINED_END_TO_END_SUMMARY={
    "final_update_global_cosine_record_only":combined_update[
        "global_cosine"
    ],
    "final_update_relative_error_record_only":combined_update[
        "relative_error"
    ],
    "min_material_parameter_update_cosine_record_only":combined_update[
        "min_material_parameter_cosine"
    ],
}

SUPPORT_FORMAL_PASS=bool(
    R5C1_PRECHECK["passed"]
    and HETERO_NATIVE_SHV_SUMMARY["passed"]
    and HETERO_NATIVE_CONTROL_SUMMARY["passed"]
)

for name,obj in [
    ("heterocoded_native_shv_summary.json",HETERO_NATIVE_SHV_SUMMARY),
    ("heterocoded_native_control_summary.json",HETERO_NATIVE_CONTROL_SUMMARY),
    ("combined_end_to_end_support_summary.json",COMBINED_END_TO_END_SUMMARY),
]:
    (OUT/name).write_text(
        json.dumps(obj,indent=2),encoding="utf-8"
    )

print(json.dumps(FIRST_STEP_GRAD_SUMMARY,indent=2))
print(json.dumps(HETERO_NATIVE_SHV_SUMMARY,indent=2))
print(json.dumps(HETERO_NATIVE_CONTROL_SUMMARY,indent=2))
print(json.dumps(COMBINED_END_TO_END_SUMMARY,indent=2))
```

```python
@torch.no_grad()
def evaluate_state_loss(state,panel,phase):
    name_to_param=dict(unique_named_parameters())

    for name,value in state.items():
        name_to_param[name].copy_(
            value.to(
                device=name_to_param[name].device,
                dtype=name_to_param[name].dtype,
            )
        )

    rows=[]

    for domain in ["language","code"]:
        seqs=panel[domain]

        for start in range(0,len(seqs),TRAIN_BATCH_SIZE):
            batch=torch.stack(
                seqs[start:start+TRAIN_BATCH_SIZE],
                dim=0,
            ).to(
                DEVICE,dtype=torch.long
            )

            logits=native_full_lm_logits(batch)
            loss=causal_lm_loss(logits,batch)

            rows.append({
                "phase":phase,
                "domain":domain,
                "batch_start":start,
                "batch_size":int(batch.shape[0]),
                "loss":float(loss),
            })

            del batch,logits,loss
            cleanup()

    df=pd.DataFrame(rows)

    return df,{
        "phase":phase,
        "mean_loss":float(
            np.average(df.loss,weights=df.batch_size)
        ),
        "domain_mean_loss":{
            domain:float(
                np.average(
                    df[df.domain==domain].loss,
                    weights=df[df.domain==domain].batch_size,
                )
            )
            for domain in ["language","code"]
        },
    }


def three_way_loss_summary(
    conventional,
    hetero_native,
    hetero_shv,
    max_gap,
):
    def rel(a,b):
        return float(abs(a-b)/max(abs(a),1e-12))

    gaps={
        "hetero_native_vs_hetero_shv":rel(
            hetero_native["mean_loss"],hetero_shv["mean_loss"]
        ),
        "conventional_native_vs_hetero_native":rel(
            conventional["mean_loss"],hetero_native["mean_loss"]
        ),
        "conventional_native_vs_hetero_shv":rel(
            conventional["mean_loss"],hetero_shv["mean_loss"]
        ),
    }

    return {
        "conventional_native":conventional,
        "heterocoded_native":hetero_native,
        "heterocoded_shv":hetero_shv,
        "pairwise_relative_gaps":gaps,
        "max_pairwise_relative_gap":max(gaps.values()),
        "passed":bool(max(gaps.values())<=max_gap),
    }


if SUPPORT_FORMAL_PASS:
    pc_rows,pc=evaluate_state_loss(
        CONV_NATIVE_FINAL,PANELS["promotion"],"promotion_conventional_native"
    )
    pn_rows,pn=evaluate_state_loss(
        HET_NATIVE_FINAL,PANELS["promotion"],"promotion_heterocoded_native"
    )
    ps_rows,ps=evaluate_state_loss(
        HET_SHV_FINAL,PANELS["promotion"],"promotion_heterocoded_shv"
    )

    PROMOTION_SUMMARY=three_way_loss_summary(
        pc,pn,ps,R5C1_PROMOTION_PAIRWISE_LOSS_GAP_MAX
    )

    pc_rows.to_csv(
        OUT/"promotion_conventional_native_eval.csv",index=False
    )
    pn_rows.to_csv(
        OUT/"promotion_heterocoded_native_eval.csv",index=False
    )
    ps_rows.to_csv(
        OUT/"promotion_heterocoded_shv_eval.csv",index=False
    )
else:
    PROMOTION_SUMMARY={
        "phase":"promotion",
        "sealed":True,
        "passed":False,
        "reason":"fresh Support formal gate failed",
    }


if PROMOTION_SUMMARY.get("passed",False):
    gc_rows,gc_res=evaluate_state_loss(
        CONV_NATIVE_FINAL,PANELS["guard"],"guard_conventional_native"
    )
    gn_rows,gn_res=evaluate_state_loss(
        HET_NATIVE_FINAL,PANELS["guard"],"guard_heterocoded_native"
    )
    gs_rows,gs_res=evaluate_state_loss(
        HET_SHV_FINAL,PANELS["guard"],"guard_heterocoded_shv"
    )

    GUARD_SUMMARY=three_way_loss_summary(
        gc_res,gn_res,gs_res,R5C1_GUARD_PAIRWISE_LOSS_GAP_MAX
    )

    gc_rows.to_csv(
        OUT/"guard_conventional_native_eval.csv",index=False
    )
    gn_rows.to_csv(
        OUT/"guard_heterocoded_native_eval.csv",index=False
    )
    gs_rows.to_csv(
        OUT/"guard_heterocoded_shv_eval.csv",index=False
    )
else:
    GUARD_SUMMARY={
        "phase":"guard",
        "sealed":True,
        "passed":False,
        "reason":"Promotion failed or remained sealed",
    }


(OUT/"promotion_three_way_summary.json").write_text(
    json.dumps(PROMOTION_SUMMARY,indent=2),encoding="utf-8"
)
(OUT/"guard_three_way_summary.json").write_text(
    json.dumps(GUARD_SUMMARY,indent=2),encoding="utf-8"
)

print(json.dumps(PROMOTION_SUMMARY,indent=2))
print(json.dumps(GUARD_SUMMARY,indent=2))

restore_state_cpu(INITIAL_STATE_CPU)
for p in model.parameters():
    p.grad=None
cleanup()
```

```python
def phase_peak_attribution(
    activation_kind,
    optimizer_kind,
    batch,
):
    assert activation_kind in ["native","shv"]
    assert optimizer_kind in ["conventional","heterocoded"]

    restore_state_cpu(INITIAL_STATE_CPU)
    set_all_model_param_grad_flags(True)
    for p in model.parameters():
        p.grad=None
    cleanup()

    opt=(
        make_conventional_optimizer()
        if optimizer_kind=="conventional"
        else make_heterocoded_optimizer()
    )
    state_summary=optimizer_state_byte_summary(opt)
    baseline_alloc=int(torch.cuda.memory_allocated())

    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()
    t0=time.perf_counter()
    loss=full_lm_loss(batch,activation_kind)
    torch.cuda.synchronize()
    forward_ms=(time.perf_counter()-t0)*1000.0
    forward_peak=int(torch.cuda.max_memory_allocated())

    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()
    t1=time.perf_counter()
    loss.backward()
    torch.cuda.synchronize()
    backward_ms=(time.perf_counter()-t1)*1000.0
    backward_peak=int(torch.cuda.max_memory_allocated())

    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()
    t2=time.perf_counter()
    grad_norm=torch.nn.utils.clip_grad_norm_(
        all_unique_parameters(),GRAD_CLIP_NORM
    )
    torch.cuda.synchronize()
    clip_ms=(time.perf_counter()-t2)*1000.0
    clip_peak=int(torch.cuda.max_memory_allocated())

    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()
    t3=time.perf_counter()
    opt.step()
    torch.cuda.synchronize()
    optimizer_ms=(time.perf_counter()-t3)*1000.0
    optimizer_peak=int(torch.cuda.max_memory_allocated())

    peaks={
        "forward":forward_peak,
        "backward":backward_peak,
        "clip":clip_peak,
        "optimizer":optimizer_peak,
    }

    step_ms=forward_ms+backward_ms+clip_ms+optimizer_ms
    predicted_tokens=int(batch.shape[0]*(batch.shape[1]-1))

    result={
        "activation_kind":activation_kind,
        "optimizer_kind":optimizer_kind,
        "optimizer_state_bytes":state_summary["total_bytes"],
        "baseline_allocated":baseline_alloc,
        "forward_peak":forward_peak,
        "backward_peak":backward_peak,
        "clip_peak":clip_peak,
        "optimizer_peak":optimizer_peak,
        "absolute_peak_allocated":max(peaks.values()),
        "dominant_peak_phase":max(peaks,key=peaks.get),
        "step_ms":step_ms,
        "predicted_tokens_per_second":float(
            predicted_tokens/(step_ms/1000.0)
        ),
        "finite":bool(
            model_parameters_are_finite()
            and optimizer_state_is_finite(opt)
        ),
    }

    del loss,grad_norm,opt
    for p in model.parameters():
        p.grad=None
    restore_state_cpu(INITIAL_STATE_CPU)
    cleanup()
    return result


MEMORY_CONFIGS=[
    ("native","conventional"),
    ("shv","heterocoded"),
]

MEMORY_ROWS=[]

for a,o in MEMORY_CONFIGS:
    print("memory warmup",a,o)
    _=phase_peak_attribution(a,o,FIXED_PROBE_BATCH)

    for rep in range(TRAIN_MEMORY_REPEATS):
        print("memory measured",a,o,rep)
        row=phase_peak_attribution(a,o,FIXED_PROBE_BATCH)
        row["repeat"]=rep
        MEMORY_ROWS.append(row)

MEMORY_DF=pd.DataFrame(MEMORY_ROWS)
MEMORY_DF.to_csv(
    OUT/"whole_training_optimizer_state_shv_benchmark.csv",
    index=False,
)

def med(a,o,col):
    frame=MEMORY_DF[
        (MEMORY_DF.activation_kind==a)&
        (MEMORY_DF.optimizer_kind==o)
    ]
    return float(frame[col].median())

native_peak=med(
    "native","conventional","absolute_peak_allocated"
)
candidate_peak=med(
    "shv","heterocoded","absolute_peak_allocated"
)
reduction=1.0-candidate_peak/native_peak

native_ms=med(
    "native","conventional","step_ms"
)
candidate_ms=med(
    "shv","heterocoded","step_ms"
)

MEMORY_SUMMARY={
    "native_conventional_peak":native_peak,
    "combined_candidate_peak":candidate_peak,
    "absolute_saved_bytes":float(native_peak-candidate_peak),
    "absolute_saved_mib":float(
        (native_peak-candidate_peak)/(1024**2)
    ),
    "whole_training_peak_reduction":float(reduction),
    "native_dominant_phase":str(
        MEMORY_DF[
            (MEMORY_DF.activation_kind=="native")&
            (MEMORY_DF.optimizer_kind=="conventional")
        ].dominant_peak_phase.mode().iloc[0]
    ),
    "candidate_dominant_phase":str(
        MEMORY_DF[
            (MEMORY_DF.activation_kind=="shv")&
            (MEMORY_DF.optimizer_kind=="heterocoded")
        ].dominant_peak_phase.mode().iloc[0]
    ),
    "native_step_ms":native_ms,
    "candidate_step_ms":candidate_ms,
    "throughput_ratio_native_over_candidate":float(
        native_ms/candidate_ms
    ),
    "candidate_predicted_tokens_per_second":med(
        "shv","heterocoded","predicted_tokens_per_second"
    ),
    "all_finite":bool(MEMORY_DF.finite.all()),
}

MEMORY_SUMMARY["passed"]=bool(
    MEMORY_SUMMARY["whole_training_peak_reduction"]>=
    R5C1_WHOLE_TRAINING_PEAK_REDUCTION_MIN
    and MEMORY_SUMMARY["all_finite"]
)

(OUT/"whole_training_optimizer_state_shv_summary.json").write_text(
    json.dumps(MEMORY_SUMMARY,indent=2),encoding="utf-8"
)

print(json.dumps(MEMORY_SUMMARY,indent=2))
```

```python

R5F0DD3_PROBE_GAP_MAX=R5C1_NATIVE_CONTROL_PROBE_GAP_MAX
R5F0DD3_UPDATE_COS_MIN=R5C1_NATIVE_CONTROL_UPDATE_COS_MIN
R5F0DD3_UPDATE_RELERR_MAX=R5C1_NATIVE_CONTROL_UPDATE_RELERR_MAX
R5F0DD3_STATE_RATIO_MAX=R5C1_EXPECTED_STATE_RATIO_MAX

class ProjectedMomentOracleAdamW:
    def __init__(self,params,mode,group_size=256,chunk_elems=MOMENT_HETEROCODE_CHUNK_ELEMS):
        self.params=list(params)
        self.mode=mode
        self.group_size=int(group_size)
        self.chunk_elems=int(chunk_elems)
        assert self.chunk_elems%self.group_size==0
        self.step_count=0
        self.pair=[]; self.qv=[]; self.qr=[]
        self.v_log_min=[]; self.v_log_step=[]; self.r_scale=[]
        self.m_exact=[]; self.v_exact=[]
        sdtype=torch.float32 if mode==PROJECTED_BOTH_FP32_SIDECAR else torch.float16
        for p in self.params:
            n=p.numel(); ng=(n+self.group_size-1)//self.group_size
            if mode in [PROJECTED_BOTH_FP16_SIDECAR,PROJECTED_BOTH_FP32_SIDECAR]:
                self.pair.append(torch.zeros((n,2),device=p.device,dtype=torch.uint8))
                self.v_log_min.append(torch.zeros(ng,device=p.device,dtype=sdtype))
                self.v_log_step.append(torch.zeros(ng,device=p.device,dtype=sdtype))
                self.r_scale.append(torch.zeros(ng,device=p.device,dtype=sdtype))
            elif mode==PROJECTED_V_ONLY:
                self.qv.append(torch.zeros(n,device=p.device,dtype=torch.uint8))
                self.v_log_min.append(torch.zeros(ng,device=p.device,dtype=torch.float16))
                self.v_log_step.append(torch.zeros(ng,device=p.device,dtype=torch.float16))
                self.m_exact.append(torch.zeros_like(p,dtype=torch.float16))
            elif mode==PROJECTED_RATIO_ONLY:
                self.qr.append(torch.full((n,),128,device=p.device,dtype=torch.uint8))
                self.r_scale.append(torch.zeros(ng,device=p.device,dtype=torch.float16))
                self.v_exact.append(torch.zeros_like(p,dtype=torch.bfloat16))
            else:
                raise ValueError(mode)
        self.last_metrics=None

    def zero_grad(self,set_to_none=True):
        for p in self.params:
            if set_to_none: p.grad=None
            elif p.grad is not None: p.grad.zero_()

    def _encode_v(self,i,start,end,v_target,qdst):
        G=self.group_size; L=end-start; ng=(L+G-1)//G; P=ng*G; g0=start//G
        pad=torch.zeros(P,device=v_target.device,dtype=torch.float32); pad[:L]=v_target
        v2=pad.view(ng,G); pos=v2>0; lv=torch.log2(v2.clamp_min(1e-30))
        lo_raw=torch.where(pos,lv,torch.full_like(lv,float("inf"))).amin(dim=1)
        hi_raw=torch.where(pos,lv,torch.full_like(lv,float("-inf"))).amax(dim=1)
        has=pos.any(dim=1); lo=torch.where(has,lo_raw,torch.zeros_like(lo_raw)); hi=torch.where(has,hi_raw,lo)
        span=(hi-lo).clamp_min(0.0); st=torch.where(span>0,span/254.0,torch.zeros_like(span))
        safe=torch.where(st>0,st,torch.ones_like(st))
        qpos=1+torch.round((lv-lo.unsqueeze(1))/safe.unsqueeze(1)).clamp(0,254)
        q=torch.where(pos,qpos,torch.zeros_like(qpos)).to(torch.uint8)
        self.v_log_min[i][g0:g0+ng].copy_(lo.to(self.v_log_min[i].dtype))
        self.v_log_step[i][g0:g0+ng].copy_(st.to(self.v_log_step[i].dtype))
        qdst[start:end].copy_(q.reshape(-1)[:L])

    def _decode_v(self,i,start,end,qsrc):
        G=self.group_size; L=end-start; ng=(L+G-1)//G; P=ng*G; g0=start//G
        flat=torch.zeros(P,device=qsrc.device,dtype=torch.int16); flat[:L]=qsrc[start:end].to(torch.int16)
        q=flat.view(ng,G)
        lo=self.v_log_min[i][g0:g0+ng].float().unsqueeze(1)
        st=self.v_log_step[i][g0:g0+ng].float().unsqueeze(1)
        pos=q>0; lv=lo+(q.float()-1.0)*st
        return torch.where(pos,torch.exp2(lv),torch.zeros_like(lv)).reshape(-1)[:L]

    def _encode_ratio(self,i,start,end,m_target,v_target,qdst):
        G=self.group_size; L=end-start; ng=(L+G-1)//G; P=ng*G; g0=start//G
        mp=torch.zeros(P,device=m_target.device,dtype=torch.float32); vp=torch.zeros(P,device=v_target.device,dtype=torch.float32)
        mp[:L]=m_target; vp[:L]=v_target
        m2=mp.view(ng,G); v2=vp.view(ng,G)
        ratio=m2/(v2.sqrt()+ADAMW_EPS)
        ma=ratio.abs().amax(dim=1); rs=torch.where(ma>0,ma/127.0,torch.zeros_like(ma))
        safe=torch.where(rs>0,rs,torch.ones_like(rs))
        qr=torch.round(ratio/safe.unsqueeze(1)).clamp(-127,127)
        qr=torch.where(rs.unsqueeze(1)>0,qr,torch.zeros_like(qr)).to(torch.int16)
        qu=(qr+128).to(torch.uint8)
        self.r_scale[i][g0:g0+ng].copy_(rs.to(self.r_scale[i].dtype))
        qdst[start:end].copy_(qu.reshape(-1)[:L])

    def _decode_ratio(self,i,start,end,qsrc,v_use):
        G=self.group_size; L=end-start; ng=(L+G-1)//G; P=ng*G; g0=start//G
        flat=torch.zeros(P,device=qsrc.device,dtype=torch.int16); flat[:L]=qsrc[start:end].to(torch.int16)-128
        q=flat.view(ng,G); rs=self.r_scale[i][g0:g0+ng].float().unsqueeze(1)
        r=(q.float()*rs).reshape(-1)[:L]
        return r*(v_use.sqrt()+ADAMW_EPS)

    @torch.no_grad()
    def step(self):
        self.step_count+=1
        beta1,beta2=ADAMW_BETAS
        bc1=1.0-beta1**self.step_count; bc2=1.0-beta2**self.step_count
        step_size=ADAMW_LR/bc1; sqrt_bc2=math.sqrt(bc2)
        ss={"m_se":0.0,"m_ref":0.0,"v_se":0.0,"v_ref":0.0,"v_zero":0}
        for i,p in enumerate(self.params):
            if p.grad is None: continue
            pf=p.view(-1); gf=p.grad.detach().view(-1)
            for start in range(0,pf.numel(),self.chunk_elems):
                end=min(start+self.chunk_elems,pf.numel())
                if self.mode in [PROJECTED_BOTH_FP16_SIDECAR,PROJECTED_BOTH_FP32_SIDECAR]:
                    qv=self.pair[i][:,0]; qr=self.pair[i][:,1]
                    old_v=self._decode_v(i,start,end,qv); old_m=self._decode_ratio(i,start,end,qr,old_v)
                elif self.mode==PROJECTED_V_ONLY:
                    old_m=self.m_exact[i].view(-1)[start:end].float()
                    old_v=self._decode_v(i,start,end,self.qv[i])
                else:
                    old_v=self.v_exact[i].view(-1)[start:end].float()
                    old_m=self._decode_ratio(i,start,end,self.qr[i],old_v)

                g=gf[start:end].float()
                m_raw=(old_m*beta1).add_(g,alpha=1.0-beta1)
                v_raw=(old_v*beta2).addcmul_(g,g,value=1.0-beta2)

                # Parent-state-lattice projection under test.
                m_target=m_raw.to(torch.float16).float()
                v_target=v_raw.to(torch.bfloat16).float()

                if self.mode in [PROJECTED_BOTH_FP16_SIDECAR,PROJECTED_BOTH_FP32_SIDECAR]:
                    qv=self.pair[i][:,0]; qr=self.pair[i][:,1]
                    self._encode_v(i,start,end,v_target,qv); v_use=self._decode_v(i,start,end,qv)
                    self._encode_ratio(i,start,end,m_target,v_target,qr); m_use=self._decode_ratio(i,start,end,qr,v_use)
                elif self.mode==PROJECTED_V_ONLY:
                    self.m_exact[i].view(-1)[start:end].copy_(m_target.to(torch.float16))
                    self._encode_v(i,start,end,v_target,self.qv[i])
                    m_use=self.m_exact[i].view(-1)[start:end].float(); v_use=self._decode_v(i,start,end,self.qv[i])
                else:
                    self.v_exact[i].view(-1)[start:end].copy_(v_target.to(torch.bfloat16))
                    v_use=self.v_exact[i].view(-1)[start:end].float()
                    self._encode_ratio(i,start,end,m_target,v_target,self.qr[i])
                    m_use=self._decode_ratio(i,start,end,self.qr[i],v_use)

                dm=m_use-m_target; dv=v_use-v_target
                ss["m_se"]+=float((dm*dm).sum()); ss["m_ref"]+=float((m_target*m_target).sum())
                ss["v_se"]+=float((dv*dv).sum()); ss["v_ref"]+=float((v_target*v_target).sum())
                ss["v_zero"]+=int(((v_target>0)&(v_use==0)).sum())

                denom=(v_use.sqrt()/sqrt_bc2).add_(ADAMW_EPS)
                target=pf[start:end].float()
                if ADAMW_WEIGHT_DECAY!=0.0: target.mul_(1.0-ADAMW_LR*ADAMW_WEIGHT_DECAY)
                target.addcdiv_(m_use,denom,value=-step_size)
                pf[start:end].copy_(target.to(pf.dtype))
        self.last_metrics={
            "m_nrmse":float((ss["m_se"]/max(ss["m_ref"],1e-30))**0.5),
            "v_nrmse":float((ss["v_se"]/max(ss["v_ref"],1e-30))**0.5),
            "positive_v_to_zero_count":int(ss["v_zero"]),
        }

    def state_tensors(self):
        out=[]
        if self.mode in [PROJECTED_BOTH_FP16_SIDECAR,PROJECTED_BOTH_FP32_SIDECAR]:
            for i in range(len(self.params)):
                out.extend([self.pair[i],self.v_log_min[i],self.v_log_step[i],self.r_scale[i]])
        elif self.mode==PROJECTED_V_ONLY:
            for i in range(len(self.params)):
                out.extend([self.m_exact[i],self.qv[i],self.v_log_min[i],self.v_log_step[i]])
        else:
            for i in range(len(self.params)):
                out.extend([self.v_exact[i],self.qr[i],self.r_scale[i]])
        return out

def make_projected_optimizer(mode):
    return ProjectedMomentOracleAdamW(all_unique_parameters(),mode,R5F0DD3_GROUP_SIZE,MOMENT_HETEROCODE_CHUNK_ELEMS)
```

```python

@torch.no_grad()
def _pack_binary_bits(bits_bool):
    bits = bits_bool.to(torch.uint8).reshape(-1)
    L = bits.numel()
    P = ((L + 7) // 8) * 8
    if P != L:
        pad = torch.zeros(
            P - L,
            device=bits.device,
            dtype=torch.uint8,
        )
        bits = torch.cat([bits, pad], dim=0)
    b = bits.view(-1, 8).to(torch.int16)
    weights = (
        1 << torch.arange(
            8,
            device=bits.device,
            dtype=torch.int16,
        )
    ).view(1, 8)
    return (b * weights).sum(dim=1).to(torch.uint8)


@torch.no_grad()
def _unpack_binary_bits(packed, length):
    x = packed.to(torch.int16).reshape(-1, 1)
    shifts = torch.arange(
        8,
        device=packed.device,
        dtype=torch.int16,
    ).view(1, 8)
    bits = ((x >> shifts) & 1).to(torch.bool).reshape(-1)
    return bits[:length]


@torch.no_grad()
def d9_reference_group_bias_encode(residual, group_size):
    G = int(group_size)
    L = residual.numel()
    ng = (L + G - 1) // G
    P = ng * G

    pad = torch.zeros(P, device=residual.device, dtype=torch.float32)
    valid = torch.zeros(P, device=residual.device, dtype=torch.float32)
    pad[:L] = residual.float()
    valid[:L] = 1.0

    r2 = pad.view(ng, G)
    v2 = valid.view(ng, G)
    count = v2.sum(dim=1).clamp_min(1.0)
    bias = (r2 * v2).sum(dim=1) / count
    return bias.to(torch.float16)


@torch.no_grad()
def d9_reference_binary_encode(residual, group_size):
    G = int(group_size)
    L = residual.numel()
    ng = (L + G - 1) // G
    P = ng * G

    pad = torch.zeros(P, device=residual.device, dtype=torch.float32)
    valid = torch.zeros(P, device=residual.device, dtype=torch.bool)
    pad[:L] = residual.float()
    valid[:L] = True

    r2 = pad.view(ng, G)
    valid2 = valid.view(ng, G)

    pos = (r2 > 0) & valid2
    neg = (~pos) & valid2

    pos_count = pos.sum(dim=1).clamp_min(1)
    neg_count = neg.sum(dim=1).clamp_min(1)

    c1 = torch.where(
        pos.any(dim=1),
        (r2 * pos).sum(dim=1) / pos_count,
        torch.zeros(ng, device=residual.device),
    )
    c0 = torch.where(
        neg.any(dim=1),
        (r2 * neg).sum(dim=1) / neg_count,
        torch.zeros(ng, device=residual.device),
    )

    bits = pos.reshape(-1)[:L]
    packed = _pack_binary_bits(bits)
    centroids = torch.stack([c0, c1], dim=1).to(torch.float16)
    return packed, centroids


@torch.no_grad()
def d9_reference_binary_decode(packed, centroids, length, group_size):
    G = int(group_size)
    bits = _unpack_binary_bits(packed, length)
    idx = torch.arange(length, device=packed.device)
    g = idx // G
    c = centroids.float()
    return torch.where(bits, c[g, 1], c[g, 0])


# Small real-code codec tests used as generator/implementation invariants.
_d9_test_r = torch.tensor(
    [-1.0, -0.5, 0.0, 0.25, 0.75, 1.0, -0.1, 0.2, 0.0],
    device=DEVICE,
    dtype=torch.float32,
)
_d9_pack, _d9_cent = d9_reference_binary_encode(_d9_test_r, 4)
_d9_dec = d9_reference_binary_decode(
    _d9_pack, _d9_cent, _d9_test_r.numel(), 4
)
assert _d9_dec.shape == _d9_test_r.shape
assert torch.isfinite(_d9_dec).all()
assert torch.equal(
    _unpack_binary_bits(
        _pack_binary_bits(_d9_test_r > 0),
        _d9_test_r.numel(),
    ),
    _d9_test_r > 0,
)
del _d9_test_r, _d9_pack, _d9_cent, _d9_dec
cleanup()
print("D9 residual codec invariants PASS")
```

```python

class MHistoryResidualAdamW(ProjectedMomentOracleAdamW):
    """
    Deployable first-moment history repair over the existing projected ratio8/log-v8 carrier.

    Persistent deployment state:
      - base ratio8/log-v8 carrier;
      - optional group-bias FP16 residual, OR
      - packed 1-bit residual indices + two FP16 centroids/group.

    Diagnostic exact-m shadow is optional and is NOT allocated in the physical memory ruler.
    """
    def __init__(
        self,
        params,
        residual_mode,
        diagnostic_shadow=False,
    ):
        super().__init__(
            params,
            mode=PROJECTED_BOTH_FP16_SIDECAR,
            group_size=R5F0DD3_GROUP_SIZE,
            chunk_elems=MOMENT_HETEROCODE_CHUNK_ELEMS,
        )
        assert residual_mode in R5F0DD9_MODES
        self.residual_mode = residual_mode
        self.diagnostic_shadow = bool(diagnostic_shadow)
        self.step_count = 0

        self.m_bias = []
        self.m_binary_bits = []
        self.m_binary_centroids = []

        for p in self.params:
            n = p.numel()
            ng = (n + self.group_size - 1) // self.group_size

            if residual_mode == GROUP_BIAS_FP16:
                self.m_bias.append(
                    torch.zeros(
                        ng,
                        device=p.device,
                        dtype=torch.float16,
                    )
                )
                self.m_binary_bits.append(None)
                self.m_binary_centroids.append(None)

            elif residual_mode == BINARY1_TWO_CENTROID:
                self.m_bias.append(None)
                self.m_binary_bits.append(
                    torch.zeros(
                        (n + 7) // 8,
                        device=p.device,
                        dtype=torch.uint8,
                    )
                )
                self.m_binary_centroids.append(
                    torch.zeros(
                        (ng, 2),
                        device=p.device,
                        dtype=torch.float16,
                    )
                )

            else:
                self.m_bias.append(None)
                self.m_binary_bits.append(None)
                self.m_binary_centroids.append(None)

        need_shadow = (
            residual_mode == EXACT_M_ORACLE
            or self.diagnostic_shadow
        )
        self.shadow_m = (
            [
                torch.zeros_like(
                    p,
                    dtype=torch.float16,
                )
                for p in self.params
            ]
            if need_shadow
            else None
        )

        self.last_metrics = None

    def _decode_group_bias(self, i, start, end):
        G = self.group_size
        L = end - start
        ng = (L + G - 1) // G
        g0 = start // G
        return (
            self.m_bias[i][g0:g0 + ng]
            .float()
            .unsqueeze(1)
            .expand(ng, G)
            .reshape(-1)[:L]
        )

    @torch.no_grad()
    def _encode_group_bias(self, i, start, end, residual):
        G = self.group_size
        L = end - start
        ng = (L + G - 1) // G
        P = ng * G
        g0 = start // G

        pad = torch.zeros(
            P,
            device=residual.device,
            dtype=torch.float32,
        )
        valid = torch.zeros(
            P,
            device=residual.device,
            dtype=torch.float32,
        )
        pad[:L] = residual.float()
        valid[:L] = 1.0

        r2 = pad.view(ng, G)
        v2 = valid.view(ng, G)
        count = v2.sum(dim=1).clamp_min(1.0)
        bias = (r2 * v2).sum(dim=1) / count

        self.m_bias[i][g0:g0 + ng].copy_(
            bias.to(torch.float16)
        )

    def _decode_binary(self, i, start, end):
        assert start % 8 == 0
        G = self.group_size
        L = end - start
        ng = (L + G - 1) // G
        g0 = start // G

        b0 = start // 8
        nb = (L + 7) // 8
        packed = self.m_binary_bits[i][b0:b0 + nb]
        bits = _unpack_binary_bits(packed, L)

        idx = torch.arange(
            L,
            device=packed.device,
            dtype=torch.long,
        )
        local_g = idx // G
        cent = self.m_binary_centroids[i][g0:g0 + ng].float()

        return torch.where(
            bits,
            cent[local_g, 1],
            cent[local_g, 0],
        )

    @torch.no_grad()
    def _encode_binary(self, i, start, end, residual):
        assert start % 8 == 0
        G = self.group_size
        L = end - start
        ng = (L + G - 1) // G
        P = ng * G
        g0 = start // G

        pad = torch.zeros(
            P,
            device=residual.device,
            dtype=torch.float32,
        )
        valid = torch.zeros(
            P,
            device=residual.device,
            dtype=torch.bool,
        )
        pad[:L] = residual.float()
        valid[:L] = True

        r2 = pad.view(ng, G)
        valid2 = valid.view(ng, G)

        pos = (r2 > 0) & valid2
        neg = (~pos) & valid2

        pos_count = pos.sum(dim=1).clamp_min(1)
        neg_count = neg.sum(dim=1).clamp_min(1)

        c1 = torch.where(
            pos.any(dim=1),
            (r2 * pos).sum(dim=1) / pos_count,
            torch.zeros(ng, device=residual.device),
        )
        c0 = torch.where(
            neg.any(dim=1),
            (r2 * neg).sum(dim=1) / neg_count,
            torch.zeros(ng, device=residual.device),
        )

        bits = pos.reshape(-1)[:L]
        packed = _pack_binary_bits(bits)

        b0 = start // 8
        self.m_binary_bits[i][b0:b0 + packed.numel()].copy_(packed)
        self.m_binary_centroids[i][g0:g0 + ng].copy_(
            torch.stack([c0, c1], dim=1).to(torch.float16)
        )

    def _decode_residual(self, i, start, end):
        if self.residual_mode == GROUP_BIAS_FP16:
            return self._decode_group_bias(i, start, end)
        if self.residual_mode == BINARY1_TWO_CENTROID:
            return self._decode_binary(i, start, end)
        return torch.zeros(
            end - start,
            device=self.params[i].device,
            dtype=torch.float32,
        )

    @torch.no_grad()
    def _encode_residual(self, i, start, end, residual):
        if self.residual_mode == GROUP_BIAS_FP16:
            self._encode_group_bias(i, start, end, residual)
        elif self.residual_mode == BINARY1_TWO_CENTROID:
            self._encode_binary(i, start, end, residual)

    @torch.no_grad()
    def step(self):
        self.step_count += 1

        beta1, beta2 = ADAMW_BETAS
        bc1 = 1.0 - beta1 ** self.step_count
        bc2 = 1.0 - beta2 ** self.step_count
        step_size = ADAMW_LR / bc1
        sqrt_bc2 = math.sqrt(bc2)

        hist_m_se = hist_m_ref = 0.0
        residual_se = residual_ref = 0.0
        v_zero = 0

        for i, p in enumerate(self.params):
            if p.grad is None:
                continue

            pf = p.view(-1)
            gf = p.grad.detach().view(-1)
            qv = self.pair[i][:, 0]
            qr = self.pair[i][:, 1]

            for start in range(0, pf.numel(), self.chunk_elems):
                end = min(start + self.chunk_elems, pf.numel())

                comp_old_v = self._decode_v(i, start, end, qv)
                comp_old_m = self._decode_ratio(
                    i, start, end, qr, comp_old_v
                )

                if self.residual_mode == EXACT_M_ORACLE:
                    old_m = (
                        self.shadow_m[i]
                        .view(-1)[start:end]
                        .float()
                    )
                elif self.residual_mode in [
                    GROUP_BIAS_FP16,
                    BINARY1_TWO_CENTROID,
                ]:
                    old_m = (
                        comp_old_m
                        + self._decode_residual(i, start, end)
                    )
                else:
                    old_m = comp_old_m

                old_v = comp_old_v
                g = gf[start:end].float()

                # Optional exact-m diagnostic shadow follows the exact first-moment recurrence
                # under the same actual candidate gradient sequence.
                exact_old_m = None
                exact_m_target = None
                if self.shadow_m is not None:
                    exact_old_m = (
                        self.shadow_m[i]
                        .view(-1)[start:end]
                        .float()
                    )
                    exact_m_raw = (exact_old_m * beta1).add(
                        g,
                        alpha=1.0 - beta1,
                    )
                    exact_m_target = (
                        exact_m_raw
                        .to(torch.float16)
                        .float()
                    )
                    self.shadow_m[i].view(-1)[start:end].copy_(
                        exact_m_target.to(torch.float16)
                    )

                    dm_hist = old_m - exact_old_m
                    hist_m_se += float((dm_hist * dm_hist).sum())
                    hist_m_ref += float(
                        (exact_old_m * exact_old_m).sum()
                    )

                m_raw = (old_m * beta1).add(
                    g,
                    alpha=1.0 - beta1,
                )
                v_raw = (old_v * beta2).addcmul(
                    g,
                    g,
                    value=1.0 - beta2,
                )

                m_target = m_raw.to(torch.float16).float()
                v_target = v_raw.to(torch.bfloat16).float()

                # Persist unchanged projected ratio8 + log-v8 base carrier.
                self._encode_v(i, start, end, v_target, qv)
                v_dec = self._decode_v(i, start, end, qv)

                self._encode_ratio(
                    i,
                    start,
                    end,
                    m_target,
                    v_target,
                    qr,
                )
                m_dec = self._decode_ratio(
                    i,
                    start,
                    end,
                    qr,
                    v_dec,
                )

                v_zero += int(
                    ((v_target > 0) & (v_dec == 0)).sum()
                )

                # Error-feedback residual targets the NEXT step's m-history.
                residual_target = m_target - m_dec
                self._encode_residual(
                    i,
                    start,
                    end,
                    residual_target,
                )

                if self.residual_mode in [
                    GROUP_BIAS_FP16,
                    BINARY1_TWO_CENTROID,
                ]:
                    residual_hat = self._decode_residual(
                        i,
                        start,
                        end,
                    )
                    dr = residual_hat - residual_target
                    residual_se += float((dr * dr).sum())
                    residual_ref += float(
                        (residual_target * residual_target).sum()
                    )

                # Frozen D9 contract: current update uses exact target state.
                u_target = m_target / (
                    v_target.sqrt() / sqrt_bc2
                    + ADAMW_EPS
                )

                target = pf[start:end].float()
                if ADAMW_WEIGHT_DECAY != 0.0:
                    target.mul_(
                        1.0
                        - ADAMW_LR
                        * ADAMW_WEIGHT_DECAY
                    )
                target.add_(
                    u_target,
                    alpha=-step_size,
                )
                pf[start:end].copy_(
                    target.to(pf.dtype)
                )

        self.last_metrics = {
            "history_m_nrmse": (
                float(
                    (
                        hist_m_se
                        / max(hist_m_ref, 1e-30)
                    ) ** 0.5
                )
                if self.shadow_m is not None
                else None
            ),
            "residual_reconstruction_nrmse": (
                float(
                    (
                        residual_se
                        / max(residual_ref, 1e-30)
                    ) ** 0.5
                )
                if self.residual_mode in [
                    GROUP_BIAS_FP16,
                    BINARY1_TWO_CENTROID,
                ]
                else None
            ),
            "positive_v_to_zero_count": int(v_zero),
        }

    def state_tensors(self):
        out = super().state_tensors()

        if self.residual_mode == GROUP_BIAS_FP16:
            out.extend(self.m_bias)

        elif self.residual_mode == BINARY1_TWO_CENTROID:
            for bits, cent in zip(
                self.m_binary_bits,
                self.m_binary_centroids,
            ):
                out.extend([bits, cent])

        elif self.residual_mode == EXACT_M_ORACLE:
            # Exact-m oracle state is intentionally included as diagnostic state.
            out.extend(self.shadow_m)

        # diagnostic_shadow for deployable candidates is deliberately NOT
        # part of deployment-state accounting and is disabled in physical rulers.
        return out


def make_d9_optimizer(
    mode,
    diagnostic_shadow=False,
):
    return MHistoryResidualAdamW(
        all_unique_parameters(),
        residual_mode=mode,
        diagnostic_shadow=diagnostic_shadow,
    )
```

```python

@torch.no_grad()
def core_r0r1_restore_unique_parameter_snapshot(state):
    name_to_param = dict(unique_named_parameters())

    state_keys = set(state.keys())
    param_keys = set(name_to_param.keys())

    missing = sorted(param_keys - state_keys)
    extra = sorted(state_keys - param_keys)

    if missing or extra:
        raise RuntimeError({
            "unique_parameter_snapshot_key_mismatch": {
                "missing_count": len(missing),
                "extra_count": len(extra),
                "missing_preview": missing[:8],
                "extra_preview": extra[:8],
            }
        })

    for name, value in state.items():
        p = name_to_param[name]
        p.copy_(
            value.to(
                device=p.device,
                dtype=p.dtype,
            )
        )


core_r0r1_restore_unique_parameter_snapshot(INITIAL_UNIQUE)
_core_r0r1_roundtrip = snapshot_unique_params_cpu()

_core_r0r1_roundtrip_exact = True
_core_r0r1_roundtrip_maxabs = 0.0

for _name, _value in INITIAL_UNIQUE.items():
    _got = _core_r0r1_roundtrip[_name]
    _core_r0r1_roundtrip_exact = (
        _core_r0r1_roundtrip_exact
        and torch.equal(
            _value,
            _got,
        )
    )
    _core_r0r1_roundtrip_maxabs = max(
        _core_r0r1_roundtrip_maxabs,
        float(
            (
                _value.float()
                - _got.float()
            ).abs().max()
        ),
    )

CORE_R0R1_SNAPSHOT_RESTORE_AUDIT = {
    "snapshot_type": "unique_named_parameters",
    "roundtrip_exact": bool(
        _core_r0r1_roundtrip_exact
    ),
    "roundtrip_maxabs": float(
        _core_r0r1_roundtrip_maxabs
    ),
    "passed": bool(
        _core_r0r1_roundtrip_exact
        and _core_r0r1_roundtrip_maxabs == 0.0
    ),
}

del _core_r0r1_roundtrip
cleanup()

(CORE_R0R1_OUT / "snapshot_restore_audit.json").write_text(
    json.dumps(
        CORE_R0R1_SNAPSHOT_RESTORE_AUDIT,
        indent=2,
    ),
    encoding="utf-8",
)

print(json.dumps(
    CORE_R0R1_SNAPSHOT_RESTORE_AUDIT,
    indent=2,
))

if not CORE_R0R1_SNAPSHOT_RESTORE_AUDIT["passed"]:
    raise RuntimeError({
        "core_r0r1_snapshot_restore_contract_failed":
            CORE_R0R1_SNAPSHOT_RESTORE_AUDIT
    })
```

## R46-B — Evaluation CE tiling: *same native transformer*, smaller logits

The native transformer and final LayerNorm are identical to the existing R45 probe. Instead of materializing `[B, T, vocab]` logits followed by an FP32 copy, evaluate the readout on consecutive **token chunks** and aggregate the *sum* of token-level negative log likelihood. Numerically this should agree up to floating-point rounding; both independent probe losses are checked before any memory claim.

This is **not a new SHV encoding scheme**, and never enters the gradient/update path. Any potential improvement is an evaluation-memory engineering effect, not proof of activation-compression superiority.

```python
@torch.no_grad()
def r46_tiled_native_probe(batch,tile_tokens=16):
    assert batch.ndim==2 and batch.shape[1]==SEQ_LEN
    assert tile_tokens>0
    H=training_embedding_forward(batch)
    H=native_transformer_body(H)
    H=model.transformer.ln_f(H)
    valid_tokens=batch.shape[0]*(batch.shape[1]-1)
    total=torch.zeros((),device=batch.device,dtype=torch.float64)
    # Causal LM label at time t is the next input token (t+1).
    for start in range(0,SEQ_LEN-1,tile_tokens):
        end=min(start+tile_tokens,SEQ_LEN-1)
        logits=model.lm_head(H[:,start:end,:])
        labels=batch[:,start+1:end+1].contiguous()
        ce_sum=F.cross_entropy(logits.float().reshape(-1,logits.shape[-1]),labels.reshape(-1),reduction='sum')
        total+=ce_sum.to(torch.float64)
        del logits,labels,ce_sum
    out=float((total/valid_tokens).detach().item())
    del total,H
    return out

def r46_probe(batch,policy):
    if policy=='full':return r45_native_probe(batch)
    if policy=='tiled':return r46_tiled_native_probe(batch,R46_GATES['tile_tokens'])
    raise ValueError(policy)

def r46_check_loss_equivalence(probe_batch,diag_batch):
    restore_state_cpu(INITIAL_STATE_CPU)
    model.eval();set_all_model_param_grad_flags(True);r45_clear_grad();cleanup()
    rows=[]
    for name,b in [('training_first_batch',probe_batch),('exposed_diagnostic',diag_batch)]:
        ref=r46_probe(b,'full');til=r46_probe(b,'tiled')
        rows.append({'panel':name,'full':ref,'tiled':til,'abs_delta':abs(til-ref),
                     'rel_delta':abs(til-ref)/max(abs(ref),1e-12),
                     'pass':bool(abs(til-ref)<=R46_GATES['probe_absolute_loss_delta_max'])})
    cleanup()
    return rows

def r46_exact_parameter_digest():
    # Streaming SHA-256 of all unique parameter bit patterns (not just sampled names).
    h=hashlib.sha256()
    with torch.no_grad():
        for name,param in unique_named_parameters():
            raw=param.detach().contiguous().view(torch.uint8).cpu().numpy().tobytes()
            h.update(name.encode('utf-8'));h.update(len(raw).to_bytes(8,'little'));h.update(raw)
    return h.hexdigest()
```

## R46-C — Per-step **and per-phase** GPU memory trace

At each step capture allocated and reserved memory, and phase-local allocated peaks for forward, backward, gradient clipping, D9, and two evaluation probes. The complete trajectory peak is the maximum of all phases, **not** the optimizer-only peak. Each policy runs on the identical starting weights and uses the same optimizer and training batches. A full bitwise digest of all model parameters checks that the probe policy never changes updates.

Phase timings include synchronization and are diagnostic only; they are **not** a formal speed benchmark.

```python
def r46_run_trajectory(variant,policy,diag_batch,steps=8):
    if variant not in R46_ARMS or policy not in R46_POLICIES:raise ValueError((variant,policy))
    assert 1<=steps<=8
    restore_state_cpu(INITIAL_STATE_CPU)
    model.eval();set_all_model_param_grad_flags(True)
    r45_clear_grad();cleanup()
    opt=make_core_r0r1_optimizer()
    torch.cuda.synchronize()
    trace=[];probe_log=[]
    max_train_peak=0.0; max_probe_peak=0.0; max_all_peak=0.0
    for step,batch in enumerate(CORE_R0R1_TRAIN_BATCHES[:steps]):
        opt.zero_grad(set_to_none=True)
        torch.cuda.synchronize()
        r46_phase_reset();start=r46_memory()
        start_time=time.perf_counter()

        loss=r45_loss(batch,variant)
        after_forward=r46_memory()
        r46_phase_reset()
        loss.backward()
        after_backward=r46_memory()
        r46_phase_reset()
        grad_norm=float(torch.nn.utils.clip_grad_norm_(all_unique_parameters(),GRAD_CLIP_NORM).detach())
        after_clip=r46_memory()
        r46_phase_reset()
        opt.step()
        after_optimizer=r46_memory()
        train_msec=1000*(time.perf_counter()-start_time)
        loss_num=float(loss.detach())
        train_peak=max(after_forward['peak_mib'],after_backward['peak_mib'],after_clip['peak_mib'],after_optimizer['peak_mib'])
        max_train_peak=max(max_train_peak,train_peak)
        max_all_peak=max(max_all_peak,train_peak)

        phase_peak_a=None;phase_peak_b=None;values={}
        probe1_ms=None;probe2_ms=None
        if policy!='none':
            r46_phase_reset()
            probe_tic=time.perf_counter()
            values['in_train']=r46_probe(CORE_R0R1_PROBE,policy)
            phase_peak_a=r46_memory()
            probe1_ms=(time.perf_counter()-probe_tic)*1000
            r46_phase_reset()
            probe_tic=time.perf_counter()
            values['exposed_diag']=r46_probe(diag_batch,policy)
            phase_peak_b=r46_memory()
            probe2_ms=(time.perf_counter()-probe_tic)*1000
            max_probe_peak=max(max_probe_peak,phase_peak_a['peak_mib'],phase_peak_b['peak_mib'])
            max_all_peak=max(max_all_peak,phase_peak_a['peak_mib'],phase_peak_b['peak_mib'])
            probe_log.append({'variant':variant,'policy':policy,'step':step,**values})

        r45_clear_grad();del loss
        cleanup()
        post=r46_memory()
        trace.append({'variant':variant,'policy':policy,'step':step,'train_loss':loss_num,'preclip_norm':grad_norm,
            'train_step_ms_instrumented':train_msec,
            'probe1_ms_instrumented':probe1_ms,'probe2_ms_instrumented':probe2_ms,
            'start_live_mib':start['live_mib'],'start_reserved_mib':start['reserved_mib'],
            'forward_peak_mib':after_forward['peak_mib'],'forward_post_live_mib':after_forward['live_mib'],
            'backward_peak_mib':after_backward['peak_mib'],'backward_post_live_mib':after_backward['live_mib'],
            'clip_peak_mib':after_clip['peak_mib'],'optimizer_peak_mib':after_optimizer['peak_mib'],
            'optimizer_post_live_mib':after_optimizer['live_mib'],
            'probe1_peak_mib':phase_peak_a['peak_mib'] if phase_peak_a else None,
            'probe2_peak_mib':phase_peak_b['peak_mib'] if phase_peak_b else None,
            'post_live_mib':post['live_mib'],'post_reserved_mib':post['reserved_mib'],
            'step_train_peak_mib':train_peak,
            'step_total_peak_mib':max(train_peak,phase_peak_a['peak_mib'] if phase_peak_a else 0,phase_peak_b['peak_mib'] if phase_peak_b else 0)})
        print('[R46]',variant,policy,step,'train peak',round(train_peak,2),'total peak',round(trace[-1]['step_total_peak_mib'],2),'post live',round(post['live_mib'],2))
    end_digest=r46_exact_parameter_digest()
    summary={'variant':variant,'policy':policy,'steps':steps,'train_only_peak_mib':max_train_peak,
            'probe_phase_peak_mib':max_probe_peak if policy!='none' else None,
            'whole_trajectory_peak_mib':max_all_peak,'end_params_sha256':end_digest,
            'train_step_ms_median_instrumented':float(__import__('statistics').median(r['train_step_ms_instrumented'] for r in trace)),
            'probe_pair_ms_median_instrumented':(float(__import__('statistics').median(r['probe1_ms_instrumented']+r['probe2_ms_instrumented'] for r in trace)) if policy!='none' else None),
            'final_train_loss':trace[-1]['train_loss']}
    del opt;cleanup()
    return trace,probe_log,summary

def r46_csv(name,records):
    if not records:return
    cols=list(records[0])
    with (R46_OUT/name).open('w',newline='',encoding='utf-8') as fp:
        writer=csv.DictWriter(fp,fieldnames=cols,extrasaction='ignore')
        writer.writeheader();writer.writerows(records)
```

## R47-B — Support trial (NC, N+D9, EM4+D9)

Each trial restores the **same pretrained GPT-2 initial weights**, selects the previously fixed identical minibatch order for its seed, uses identical mixed-precision and AdamW hyperparameters, and measures CUDA training+Probe peaks separately. Probes are native **tiled CE**, only at the prespecified seven checkpoints; they do not enter backward. NC is the conventional FP16-m/BF16-v low-precision anchor, not the FP32 AdamW or mature engineered performance baseline.


```python
def r47_all_params_sha():
    return r46_exact_parameter_digest()

@torch.no_grad()
def r47_tiled_probe(batch):
    return r46_tiled_native_probe(batch, R47_GATES['tile_tokens'])


def r47_run_trial(variant,seed,panel):
    assert variant in ('NC','N','EM4')
    restore_state_cpu(INITIAL_STATE_CPU)
    model.eval();set_all_model_param_grad_flags(True)
    r45_clear_grad();cleanup()
    random.seed(seed);np.random.seed(seed);torch.manual_seed(seed);torch.cuda.manual_seed_all(seed)
    initial_hash=r47_all_params_sha()
    opt=make_conventional_optimizer() if variant=='NC' else make_core_r0r1_optimizer()
    train_trace=[];probes=[]
    max_train_peak=0.0;max_probe_peak=0.0;max_reserved=0.0
    for step,batch in enumerate(panel['train_batches']):
        opt.zero_grad(set_to_none=True)
        torch.cuda.synchronize();torch.cuda.reset_peak_memory_stats()
        t0=time.perf_counter()
        loss=(full_lm_loss(batch,'native') if variant=='NC' else r45_loss(batch, 'N' if variant=='N' else 'EM4'))
        loss_num=float(loss.detach().item())
        if not math.isfinite(loss_num):raise RuntimeError(f'nonfinite loss {variant}/{seed}/{step}')
        loss.backward()
        grad_norm=float(torch.nn.utils.clip_grad_norm_(all_unique_parameters(),GRAD_CLIP_NORM).detach())
        opt.step();torch.cuda.synchronize()
        step_ms=1000.0*(time.perf_counter()-t0)
        train_peak=torch.cuda.max_memory_allocated()/2**20
        reserved=torch.cuda.max_memory_reserved()/2**20
        max_train_peak=max(max_train_peak,train_peak);max_reserved=max(max_reserved,reserved)
        del loss
        r45_clear_grad();cleanup()
        train_trace.append({'seed':seed,'variant':variant,'step':step,'train_loss':loss_num,
            'preclip_norm':grad_norm,'train_step_ms':step_ms,'train_peak_mib':train_peak,
            'train_reserved_peak_mib':reserved})
        if step in R47_GATES['eval_after_steps']:
            torch.cuda.synchronize();torch.cuda.reset_peak_memory_stats()
            held=r47_tiled_probe(panel['diag_batch'])
            train_first=r47_tiled_probe(panel['train_first_batch'])
            torch.cuda.synchronize()
            probe_peak=torch.cuda.max_memory_allocated()/2**20
            max_reserved=max(max_reserved,torch.cuda.max_memory_reserved()/2**20)
            max_probe_peak=max(max_probe_peak,probe_peak)
            probes.append({'seed':seed,'variant':variant,'step':step,
                'train_first_probe':train_first,'fresh_validation_probe':held,'probe_peak_mib':probe_peak})
        if step in (0,7,15):print('[R47]',seed,variant,step,'loss',round(loss_num,6),'peak MiB',round(train_peak,3))
    final_hash=r47_all_params_sha()
    del opt;cleanup()
    return train_trace,probes,{'seed':seed,'variant':variant,'steps':len(train_trace),
        'initial_params_sha256':initial_hash,'final_params_sha256':final_hash,
        'train_peak_mib':max_train_peak,'probe_peak_mib':max_probe_peak,
        'whole_peak_mib':max(max_train_peak,max_probe_peak),
        'max_reserved_mib':max_reserved,
        'median_train_step_ms':float(statistics.median(t['train_step_ms'] for t in train_trace)),
        'final_loss':train_trace[-1]['train_loss']}


def r47_seed_gate(summaries,probes,seed):
    s={v:next(r for r in summaries if r['seed']==seed and r['variant']==v) for v in ('NC','N','EM4')}
    assert s['NC']['initial_params_sha256']==s['N']['initial_params_sha256']==s['EM4']['initial_params_sha256']
    def proj(v):return {x['step']:x for x in probes if x['seed']==seed and x['variant']==v}
    ref=proj('NC');cand=proj('EM4')
    assert set(ref)==set(cand)==set(R47_GATES['eval_after_steps'])
    gaps=[]
    for step in sorted(ref):
        for field in ('train_first_probe','fresh_validation_probe'):
            gap=abs(cand[step][field]-ref[step][field])/max(abs(ref[step][field]),1e-12)
            gaps.append({'seed':seed,'step':step,'field':field,'relative_gap':gap})
    max_gap=max(x['relative_gap'] for x in gaps)
    saving=1-s['EM4']['whole_peak_mib']/s['N']['whole_peak_mib']
    slow=s['EM4']['median_train_step_ms']/s['N']['median_train_step_ms']
    quality=bool(max_gap<=R47_GATES['loss_gap_vs_nc_max'])
    memory=bool(saving>=R47_GATES['matched_em4_vs_n_d9_peak_reduction_min'])
    speed=bool(slow<=R47_GATES['em4_vs_n_d9_step_time_ratio_max'])
    return {'seed':seed,'max_nc_em4_probe_rel_gap':max_gap,
        'em4_vs_n_peak_saving':saving,'em4_vs_n_step_time_ratio':slow,
        'em4_vs_nc_step_time_ratio':s['EM4']['median_train_step_ms']/s['NC']['median_train_step_ms'],
        'pass_quality':quality,'pass_memory':memory,'pass_speed_pilot':speed,
        'pass':bool(quality and memory and speed)},gaps
```

```python
def r47_package_export():
    # Always export latest data, including exception diagnostics, without changing mathematical code.
    r47_write_json('summary.json',R47_OUT_JSON)
    r47_write_json('protocol.json',R47_GATES)
    r47_write_json('environment_r47.json',{
        'gpu':torch.cuda.get_device_name(0) if torch.cuda.is_available() else None,
        'torch':torch.__version__,'cuda':torch.version.cuda,
        'transformers':transformers.__version__,'datasets':datasets.__version__,
        'model_id':MODEL_ID,'wikitext_revision':WIKITEXT_REV,
        'note':'GPU allocated/reserved + synchronized wall-clock; pilot not formal engineering benchmark'})
    files=sorted(q for q in R47_OUT.iterdir() if q.is_file())
    fmap={f.name:hashlib.sha256(f.read_bytes()).hexdigest() for f in files if f.name!='RESULT_MANIFEST.json'}
    r47_write_json('RESULT_MANIFEST.json',{'round':R47_ROUND,'file_sha256':fmap,
        'frozen_parent_code_sha256':R47_PARENT_CODE_SHA256,'timestamp_utc':datetime.now(timezone.utc).isoformat()})
    output_root=Path('/content') if Path('/content').exists() else Path('.')
    zp=output_root/'SHV_R47_FRESH_B0_RESULT_UPLOAD_TO_CHATGPT.zip'
    with zipfile.ZipFile(zp,'w',compression=zipfile.ZIP_DEFLATED,compresslevel=8) as out:
        for f in sorted(R47_OUT.iterdir()):
            if f.is_file():out.write(f,f.name)
    assert zipfile.ZipFile(zp).testzip() is None
    digest=hashlib.sha256(zp.read_bytes()).hexdigest()
    sha_file=output_root/'SHV_R47_FRESH_B0_RESULT_UPLOAD_TO_CHATGPT.zip.sha256'
    sha_file.write_text(digest+'  '+zp.name+'\n',encoding='utf-8')
    print('[R47] STATUS',R47_OUT_JSON['status'],'ZIP',str(zp),'SHA256',digest)
    try:
        from google.colab import files as colab_files
        colab_files.download(str(zp));colab_files.download(str(sha_file))
    except ImportError: pass

R47_TRACE=[];R47_PROBES=[];R47_SUMMARIES=[];R47_SEED_GATES=[];R47_GAPS=[]
try:
    r47_preflight()
    R47_PANELS,R47_DATA_AUDIT=r47_prepare_data()
    R47_OUT_JSON['freshness_audit_digest']=R47_DATA_AUDIT['all_new_hashes_sha256']
    R47_OUT_JSON['status']='RUNNING'
    for i,seed in enumerate(R47_GATES['data_seeds']):
        for variant in ('NC','N','EM4'):
            a,b,c=r47_run_trial(variant,seed,R47_PANELS[seed])
            R47_TRACE.extend(a);R47_PROBES.extend(b);R47_SUMMARIES.append(c)
        g,rows=r47_seed_gate(R47_SUMMARIES,R47_PROBES,seed)
        R47_SEED_GATES.append(g);R47_GAPS.extend(rows)
        print('[R47] seed gate',seed,json.dumps(g,ensure_ascii=False))
        # Strict prereg stop after seed 0 when its new data do not support the mechanism.
        if i==0 and not g['pass'] and R47_GATES['stop_after_seed0_if_failed']:
            R47_OUT_JSON['status']='FRESH_B0_SEED0_HOLD'; break
    else:
        R47_OUT_JSON['status']='FRESH_B0_ELIGIBLE_FOR_B1' if all(g['pass'] for g in R47_SEED_GATES) else 'FRESH_B0_HOLD'
    R47_OUT_JSON['seed_results']=R47_SEED_GATES
except Exception:
    R47_OUT_JSON['status']='EXCEPTION'
    R47_OUT_JSON['exception']=traceback.format_exc()
    print('[R47 EXCEPTION]',R47_OUT_JSON['exception'])
finally:
    r47_csv('training_step_trace.csv',R47_TRACE)
    r47_csv('probe_values.csv',R47_PROBES)
    r47_csv('run_summaries.csv',R47_SUMMARIES)
    r47_csv('seed_gates.csv',R47_SEED_GATES)
    r47_csv('relative_gap.csv',R47_GAPS)
    r47_package_export()
```
